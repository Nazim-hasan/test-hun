## **Sudoku**

Create a class called **SudokuBoard** that represents a Sudoku board. Sudoku is a 9×9 board divided into nine 3×3 blocks along the row and column boundaries. You need to place the numbers from 1 to 9 so that in every row, every column, and every 3×3 block each number appears exactly once. The internal representation of the board is up to you.

The class must provide the following member functions:

*   Create a constructor that creates a `SudokuBoard` object representing an empty board.
*   `can_place(row_idx, col_idx, num) -> bool`: Returns a boolean indicating whether the number `num` can be placed at position (`row_idx`, `col_idx`) according to the rules of Sudoku. Row and column indexing starts from 0.
*   `place(pos, num) -> bool`: Returns a boolean indicating whether the number `num` can be placed at position `pos` according to the rules of Sudoku. `pos` is a tuple containing two integers between 0 and 8. Its first element is the row index, and its second element is the column index. If the function returns `True`, it must also place the given number on the board at the given position.

*   `get_max_of_rows() -> ndarray(9,)`: For each row, returns the largest number, arranged into a 1D array of shape `(9,)`. If a row contains no numbers, return 0 at the corresponding position.

*   `sum_of_missing_numbers_per_block() -> ndarray(3, 3)`: Returns an array of shape `(3, 3)` whose elements are the sums of the missing numbers in each 3×3 block.

*   `get_row_idx_with_highest_sum() -> int`: Returns the index of the row whose placed numbers have the largest sum. If there are multiple such rows, return any one of them.
*   `get_num_cols_where_number_is_present(num) -> int`: Returns how many columns contain the number `num`.
*   `get_num_empty_rows() -> int`: Returns the number of empty rows.
*   `get_col_idxs_with_missing_num(num) -> ndarray(n_cols_ret,)`: Returns the indices of the columns that do not yet contain the number `num`, in a 1D array of shape `(n_cols_ret,)`.

*   `get_empty_block_idxs() -> ndarray(n_empty_blocks, 2)`: Returns the indices of those 3×3 blocks that do not contain any numbers, in an array of shape `(n_empty_blocks, 2)`. Each block position is given by two indices. If you view the board as a matrix, for example the bottom-left block has index `[2, 0]`.

*   `get_row_with_most_numbers() -> ndarray(9,)`: Returns the row that contains the most placed numbers, as a 1D array. If there are multiple such rows, return any one of them. Empty positions in the returned array should be indicated by 0 values.

*   `get_valid_col_idxs_to_place(num) -> ndarray(n_valid_cols,)`: Returns the indices (0-based) of the columns where the given number `num` could be placed. It is sufficient to check only whether the given column already contains `num` or not; there is no need to check uniqueness with respect to rows and 3×3 blocks.

*   `get_num_rows_with_two_nums_present(num1, num2) -> int`: Returns the number of rows in which both `num1` and `num2` are present at the same time. You may assume the function is always called with two different numbers.

*   `get_max_of_rows_where_num_is_present(num) -> int`: Takes all rows in which `num` is present and returns the maximum number within these rows. If `num` does not appear anywhere on the board, return 0.

*   `get_rightmost_numbers_in_rows() -> ndarray(9,)`: Returns, for each row, the rightmost (last by column position) placed number, arranged into an array. For empty rows, 0 values must be placed at the corresponding positions in the returned array.
*   `get_block_pos_with_minimal_max() -> (int, int) or None`: Returns the position of the non-empty 3×3 block whose maximum value is minimal among the maxima of all non-empty blocks. If there are multiple such blocks, returning any one of them is sufficient. If all blocks are empty, return `None`. The block position is specified by two indices. If you view the board as a matrix, for example the bottom-left block has position `(2, 0)`. Return the two indices as a 2-element tuple.

*   `get_cols_with_above_average_sum() -> ndarray(9, n_above_avg_cols)`: Returns the columns whose column-sum is greater than the average column-sum, stacked next to each other into a 2D array, preserving the same order and orientation of columns as in the original board. Empty positions must be indicated by 0 values in the returned array. The sum of an empty column is 0. If the board is empty, the returned array has shape `(9, 0)`.

*   `get_num_numbers_greater_than_col_avg() -> int`: Returns the number of entries on the board that are greater than the average of the numbers in their own column.

*   `get_num_numbers_smaller_than_row_avg() -> int`: Returns the number of entries on the board that are smaller than the average of the numbers in their own row.

*   `get_matching_nonempty_col_row_pair_idxs() -> ndarray(n_matching_col_row_pairs, 2)`: Finds those column-row pairs whose contents match each other (ignoring the transposed orientation). Returns the associated (column index, row index) pairs as the rows of a 2D array. The array has two columns: the first contains column indices and the second contains row indices. If the board is empty, the returned array has shape `(0, 2)`.

*   `get_number_frequencies(nums_to_query) -> ndarray(len(nums_to_query),)`: For each number listed in the `nums_to_query` argument, returns its frequency (number of occurrences) on the board. `nums_to_query` may be any `sequence` type.

*   `get_first_occurence_of_numbers() -> (ndarray(9,), ndarray(9,))`: For each number from 1 to 9, returns the 2D index of its first occurrence. The board is scanned starting from the top-left corner, proceeding row by row, elements in each row taken from left to right. The 2D indices follow the standard NumPy format: a 2-tuple of two 1D arrays of equal length, storing row indices and column indices separately. If a given number does not appear on the board, the function must return index 9 in both returned arrays for that number.

*   `get_second_largest_number_below_5_in_rows() -> ndarray(9,)`: For each row, returns the second largest number among those that are smaller than 5, arranged into an array of shape `(9,)`. If a row contains no such number (i.e., it contains at most one number smaller than 5), place 0 at the corresponding position in the returned array.

*   `get_top5_num_with_pos() -> ndarray(5, 3)`: Returns the five largest numbers found on the board and their positions, ordered by decreasing value. Along the second axis, store (in order) the number value, the row index, and the column index. If ties make it ambiguous which positions to include, you may choose any of them; only the ordering by value matters. If there are fewer than five numbers on the board, fill the remaining bottom rows of the returned array with `[0, 9, 9]` (in order).

### Solution


```python
import numpy as np

# Implement your solution here.

class SudokuBoard:

    def __init__(self):
        self.board = np.zeros((9, 9), dtype=np.uint8)
        self.board_blockview = self.board.reshape(3, 3, 3, 3)   # (n_blockrow=3, n_row_in_block=3, n_blockcol=3, n_cols_in_block=3)

    def __str__(self):
        return str(self.board)

    # can_place: OLD VERSION
    '''
    def can_place(self, row_idx, col_idx, num):
        if self.board[row_idx, col_idx] != 0:
            return False

        if np.any(self.board[row_idx, :] == num):
            return False
            
        if np.any(self.board[:, col_idx] == num):
            return False

        block_start_row_idx = (row_idx // 3)*3
        block_start_col_idx = (col_idx // 3)*3

        if np.any(self.board[block_start_row_idx:block_start_row_idx+3,\
                   block_start_col_idx:block_start_col_idx+3] == num):
            return False

        return True
    '''

    # can_place: NEW VERSION using self.block_view and some extra compacting
    def can_place(self, row_idx, col_idx, num):
        b = self.board[row_idx, col_idx] != 0
        b = b or np.any(self.board[row_idx, :] == num)
        b = b or np.any(self.board[:, col_idx] == num)
        block_row_idx, block_col_idx = row_idx // 3, col_idx // 3
        b = b or np.any(self.board_blockview[block_row_idx, :, block_col_idx, :] == num)
        return not b

    def place(self, pos, num):
        row_idx, col_idx = pos
        success = self.can_place(row_idx, col_idx, num)
        if success is True:
            self.board[row_idx, col_idx] = num
        return success

    def get_max_of_rows(self):
        return np.amax(self.board, axis=1)             # (n_rows, n_cols) -> (n_rows,)
            
    def sum_of_missing_numbers_per_block(self):
        sum_of_a_full_block = np.sum(np.arange(10))   # int (== 45)
        block_sums = np.sum(self.board_blockview, axis=(1, 3))  # (n_blockrow=3, n_blockcol=3)
        return sum_of_a_full_block - block_sums
    
    def get_row_idx_with_highest_sum(self):
        return np.argmax(np.sum(self.board, axis=1))
        
    def get_num_cols_where_number_is_present(self, num):
        cols_with_num = np.any(self.board == num, axis=0)     # np.any: (n_rows, n_cols) bool -> (n_cols,) bool
        return np.count_nonzero(cols_with_num)

    def get_num_empty_rows(self):
        #return np.count_nonzero(np.all(self.board == 0, axis=1))
        return 9 - np.count_nonzero(np.any(self.board, axis=1))      # relying on implicit (automatic) int -> bool dtype conversion

    def get_col_idxs_with_missing_num(self, num):
        cols_without_num = np.all(self.board != num, axis=0)     # np.all: (n_rows, n_cols) bool -> (n_cols,) bool
        return np.where(cols_without_num)[0]

    def get_empty_block_idxs(self):
        empty_blocks = np.all(self.board_blockview == 0, axis=(1, 3))         # (n_blockrow=3, n_blockcol=3) bool
        #return np.array(np.where(empty_blocks)).T
        return np.argwhere(empty_blocks)

    def get_row_with_most_numbers(self):
        row_counts = np.count_nonzero(self.board, axis=1)     # count_nonzero: (n_rows, n_cols) -> (n_rows,)
        row_idx_with_most_numbers = np.argmax(row_counts)
        return self.board[row_idx_with_most_numbers, :]
    
    def get_valid_col_idxs_to_place(self, num):
        # ~ -> np.logical_not(...)
        return np.where(~np.any(self.board == num, axis=0))[0]

    def get_num_rows_with_two_nums_present(self, num1, num2):
        # & -> np.logical_and(..., ...)
        rows_with_both_nums = np.any(self.board == num1, axis=1) & np.any(self.board == num2, axis=1)     # (n_rows,) bool & (n_rows,) bool
        return np.count_nonzero(rows_with_both_nums)

    # Alternative solution
    def get_num_rows_with_two_nums_present(self, num1, num2):
        # | -> np.logical_or(..., ...)
        rows_with_both_nums = np.count_nonzero((self.board == num1) | (self.board == num2), axis=1) == 2
        return np.count_nonzero(rows_with_both_nums)

    def get_max_of_rows_where_num_is_present(self, num):
        '''
        rows_mask = np.any(self.board == num, axis=1)      # (n_rows,) bool
        rows_where_num_is_present = self.board[rows_mask, :]   # (n_rows, n_cols) -> (n_rows_where_num, n_cols)
        return rows_where_num_is_present.max(initial=0)
        '''
        return self.board[np.any(self.board == num, axis=1), :].max(initial=0)    # same as above in one line...

    # Alternative solution
    def get_max_of_rows_where_num_is_present(self, num):
        rows_mask = np.any(self.board == num, axis=1)      # (n_rows,) bool
        board_copy = np.copy(self.board)
        board_copy[~rows_mask, :] = 0
        return np.amax(board_copy)

    def get_rightmost_numbers_in_rows(self):
        rightmost_number_idxs = np.argmax(self.board[:,::-1] != 0, axis=1)
        return self.board[:,::-1][np.arange(9), rightmost_number_idxs]

    def get_block_pos_with_minimal_max(self):
        block_maxes = np.amax(self.board_blockview, axis=(1, 3))  # (n_blockrow=3, n_blockcol=3)
        nonempty_coords = np.where(block_maxes != 0)         # (n_nonempty_blocks,), (n_nonempty_blocks,)
        if nonempty_coords[0].shape[0] == 0:
            return None
        else:
            nonempty_blockmaxes = block_maxes[nonempty_coords]        # (n_nonempty_blocks,)
            min_nonempty_blockmax_idx = np.argmin(nonempty_blockmaxes)
            return nonempty_coords[0][min_nonempty_blockmax_idx], nonempty_coords[1][min_nonempty_blockmax_idx]

    # Alternative solution
    def get_block_pos_with_minimal_max(self):
        block_maxes = np.amax(self.board_blockview, axis=(1, 3))  # (n_blockrow=3, n_blockcol=3)
        block_maxes[block_maxes == 0] = 10
        min_blockmax_flat_idx = np.argmin(block_maxes)         # argmin returns a flat index, indexing the 1D view of the array
        # min_blockmax_idx =  min_blockmax_flat_idx // 3, min_blockmax_flat_idx % 3
        min_blockmax_idx = np.unravel_index(min_blockmax_flat_idx, block_maxes.shape)      # converting flat idx to 2D idx to index (3, 3) array
        if block_maxes[min_blockmax_idx] == 10:
            return None
        else:
            return min_blockmax_idx

    def get_cols_with_above_average_sum(self):
        col_sums = np.sum(self.board, axis=0)         # sum: (n_rows, n_cols) -> (n_cols,)
        above_avg_col_mask = col_sums > np.mean(col_sums)    # broadcasting: (n_cols,) v scalar -> (n_cols,) v (1,) -> (n_cols,) v (n_cols,)
        return self.board[:, above_avg_col_mask]      # (n_rows, n_selected_cols)

    def get_num_numbers_greater_than_col_avg(self):
        col_avgs = np.mean(self.board, axis=0, where=self.board != 0)    # mean: (n_rows, n_cols) -> (n_cols,)
        #col_avgs[np.isnan(col_avgs)] = 0.
        col_avgs = np.nan_to_num(col_avgs, nan=0.)
        mask_of_elems_count = self.board > col_avgs       # (n_rows, n_cols) v (n_cols,) -> (n_rows, n_cols)
        return np.count_nonzero(mask_of_elems_count)

    def get_num_numbers_smaller_than_row_avg(self):
        row_avgs = np.mean(self.board, axis=1, where=self.board != 0)    # mean: (n_rows, n_cols) -> (n_rows,)
        row_avgs = np.nan_to_num(row_avgs, nan=0.)
        #mask_of_elems_count = self.board < row_avgs       # INCORRECT BROADCASTING (n_rows, n_cols) v (n_rows,) -> (n_rows, n_cols)
        #mask_of_elems_count = self.board < row_avgs.reshape(-1, 1)       # (n_rows, n_cols) v (n_rows, 1) -> (n_rows, n_cols)
        mask_of_elems_count = self.board < row_avgs[:, None]              # synonyms with row_avgs.reshape(-1, 1) in this case
        nums_present = self.board > 0                                    # (n_rows, n_cols)
        return np.count_nonzero(mask_of_elems_count & nums_present)

    # Alterantive solution
    def get_num_numbers_smaller_than_row_avg(self):
        board_copy = self.board.astype(np.float32)          # copies the board as well
        board_copy[board_copy == 0.] = np.nan
        row_avgs = np.nanmean(board_copy, axis=1)
        row_avgs = np.nan_to_num(row_avgs, nan=0.)
        mask_of_elems_count = self.board < row_avgs[:, None]             # (n_rows, n_cols) v (n_rows, 1) -> (n_rows, n_cols)
        nums_present = self.board > 0                                    # (n_rows, n_cols)
        return np.count_nonzero(mask_of_elems_count & nums_present)
        
    def get_matching_nonempty_col_row_pair_idxs(self):
        #mask_of_matching_elems3d = self.board[:, None, :] == self.board.T[:, :, None]    # (n_rows, 1, n_cols) v (n_cols, n_rows, 1) -> (n_cols=n_rows, n_rows, n_cols)
        #mask_of_matching_pairs2d = np.all(mask_of_matching_elems3d, axis=0)     # (n_rows, n_cols)
        mask_of_matching_elems3d = self.board == self.board[:, :, None]                   # (n_rows, n_cols) v (n_rows, n_cols, 1) -> (n_rows, n_cols=n_rows, n_cols)
        mask_of_matching_pairs2d = np.all(mask_of_matching_elems3d, axis=1)     # (n_rows, n_cols)
        nonempty_cols_mask = np.any(self.board > 0, axis=0)             # (n_rows, n_cols) -> (n_cols,)
        nonempty_rows_mask = np.any(self.board > 0, axis=1)             # (n_rows, n_cols) -> (n_rows,)
        mask_of_matching_pairs2d &= nonempty_cols_mask            # (n_rows, n_cols) v (n_cols,) -> (n_rows, n_cols)
        mask_of_matching_pairs2d &= nonempty_rows_mask[:,None]    # (n_rows, n_cols) v (n_rows, 1) -> (n_rows, n_cols)
        #return np.argwhere(mask_of_matching_pairs2d[:,::-1])     # (n_matching_pairs, 2:[row_idx, col_idx]) -> (n_matching_pairs, 2:[col_idx, row_idx])
        return np.argwhere(mask_of_matching_pairs2d.T)            # (n_matching_pairs, 2:[col_idx, row_idx])

    # BROADCASTING rules
    # if array vs scalar: scalar is wrapped into an array of (1,)
    # if array1.ndim != array2.ndim: array of smaller dim will be matched to the other array dim by
    #                                 adding new LEADING axes of length 1 to the smaller dim array
    # arrays now have the same dimensionality, so axes at the same positions (arr.shape[i]) are compared:
    #    if arr1.shape[i] == arr2.shape[i] == n:               then no repetitions, result_arr.shape[i] == n
    #    elif (arr1.shape[i] == 1) and (arr2.shape[i] == n):   then arr1 is repeated n times along axis #i
    #                                                                           and result_arr.shape[i] == n
    #    elif (arr1.shape[i] == n) and (arr2.shape[i] == 1):   then arr2 is repeated n times along axis #i
    #                                                                           and result_arr.shape[i] == n
    #    [elif arr1.shape[i] != arr2.shape[i]]:                then raise ValueError "operands could not be broadcast together"
    # if no error was raised: broadcasting is successful, and result_arr is the result of the operation
    #

    # BROADCASTING cases
    # (a, b) + (a, b) -> (a, b)
    # (a, b) + scalar -> (a, b) + (1,) -> (a, b) + (1, 1) -> (a, b) + (a, b) -> (a, b)
    # (a, b) + (b,)   -> (a, b) + (1, b) -> (a, b) + (a, b) -> (a, b)
    # (a, b) + (a,)   -> (a, b) + (1, a) -> b != a -> BROADCASTING FAILS
    # (a, b) + (a, 1) -> (a, b)
    # (a, 1) + (b,)   -> (a, 1) + (1, b) -> (a, b) + (a, b) -> (a, b)

    def get_number_frequencies(self, nums_to_query):
        # High memory usage
        matching_mask_3d = self.board[:,:,None] == nums_to_query      # (n_rows, n_cols, 1) v (n_nums_to_query,) -> (n_rows, n_cols, n_nums_to_query)
        return np.count_nonzero(matching_mask_3d, axis=(0, 1))        # (n_rows, n_cols, n_nums_to_query) -> (n_nums_to_query,)

    # Alternative solution
    def get_number_frequencies(self, nums_to_query):
        return np.bincount(self.board.reshape(-1), minlength=10)[nums_to_query]

    def get_first_occurence_of_numbers(self):
        first_idxs0 = np.full((9,), dtype=np.int64, fill_value=9)
        first_idxs1 = np.full((9,), dtype=np.int64, fill_value=9)
        unique_nums, first_idxs_flat = np.unique(self.board, return_index=True)
        if unique_nums[0] == 0:
            unique_nums, first_idxs_flat = unique_nums[1:], first_idxs_flat[1:]
        first_idxs_valid0, first_idxs_valid1 = np.unravel_index(first_idxs_flat, shape=self.board.shape)
        first_idxs0[unique_nums-1] = first_idxs_valid0
        first_idxs1[unique_nums-1] = first_idxs_valid1
        return first_idxs0, first_idxs1

    def get_second_largest_number_below_5_in_rows(self):
        board_copy = np.copy(self.board)
        board_copy[board_copy >= 5] = 0
        return np.sort(board_copy, axis=1)[:, -2]

    def get_top5_num_with_pos(self):
        top5_flat_idxs = np.argsort(self.board, axis=None)[-5:][::-1]        # (5,)
        top5_idxs0, top5_idxs1 = np.unravel_index(top5_flat_idxs, shape=self.board.shape)        # (5,), (5,)
        # top5_numbers = self.board.reshape(-1)[top5_flat_idxs]
        top5_numbers = self.board[top5_idxs0, top5_idxs1]                    # (5,)
        top5_nums_with_pos = np.stack([top5_numbers, top5_idxs0, top5_idxs1], axis=1)    # (5, 3)
        invalid_entries = top5_nums_with_pos[:, 0] == 0             # (5,) bool
        #top5_nums_with_pos[invalid_entries, 1:] = [9, 9]           # (n_invalid_rows, 2) v (2,)   -> (n_invalid_rows, 2) v (1, 2) -> (n_invalid_rows, 2)
        top5_nums_with_pos[invalid_entries, 1:] = 9                 # (n_invalid_rows, 2) v scalar -> (n_invalid_rows, 2) v (1, 1) -> (n_invalid_rows, 2)
        return top5_nums_with_pos
        





```

```python
import unittest

class TestSudoku(unittest.TestCase):
    def test_can_place(self):
        sb = SudokuBoard()
        self.assertTrue(sb.can_place(0, 0, 5))
        self.assertTrue(sb.can_place(8, 0, 5))
        self.assertTrue(sb.can_place(0, 8, 5))
        self.assertTrue(sb.can_place(4, 6, 5))
        self.assertTrue(sb.can_place(8, 8, 5))

        sb.place((3, 3), 5)
        self.assertFalse(sb.can_place(3, 3, 7))
        self.assertFalse(sb.can_place(5, 4, 5))
        self.assertFalse(sb.can_place(7, 3, 5))
        self.assertFalse(sb.can_place(3, 7, 5))

        self.assertTrue(sb.can_place(2, 4, 5))
        self.assertTrue(sb.can_place(3, 4, 6))

    def test_place(self):
        sb = SudokuBoard()
        self.assertTrue(sb.place((2, 7), 4))
        self.assertFalse(sb.place((2, 7), 4))
        self.assertFalse(sb.place((2, 7), 5))
        self.assertTrue(sb.place((2, 0), 5))
        self.assertFalse(sb.place((2, 8), 5))

    def test_get_max_of_rows(self):
        sb = SudokuBoard()
        maxes = sb.get_max_of_rows()
        self.assertEqual(maxes.shape, (9,))
        self.assertTrue(np.all(maxes == [0]*9))

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        maxes = sb.get_max_of_rows()
        self.assertEqual(maxes.shape, (9,))
        self.assertTrue(np.all(maxes == [0,3,4,8,0,4,0,0,4]))

        to_insert = [((1, 4), 9), ((3, 2), 6), ((3, 3), 7), ((8, 5), 1), ((8, 8), 2)]
        [sb.place(*t) for t in to_insert]
        maxes = sb.get_max_of_rows()
        self.assertEqual(maxes.shape, (9,))
        self.assertTrue(np.all(maxes == [0,9,4,8,0,4,0,0,4]))

    def test_sum_of_missing_numbers_per_block(self):
        sb = SudokuBoard()
        sum_missing = sb.sum_of_missing_numbers_per_block()
        self.assertTrue(np.all(sum_missing == 45))
        self.assertEqual(sum_missing.shape, (3,3))

        to_insert = [((6,5), 6),((8,5), 7), ((7,5), 8)]
        [sb.place(*t) for t in to_insert]
        sum_missing = sb.sum_of_missing_numbers_per_block()
        self.assertEqual(sum_missing[2,1], 24)
        self.assertEqual(np.sum(sum_missing == 45), 8)

        to_insert = [((3,0), 1),((4,2), 2), ((4,1), 3)]
        [sb.place(*t) for t in to_insert]
        sum_missing = sb.sum_of_missing_numbers_per_block()
        self.assertEqual(sum_missing[1,0], 39)
        self.assertEqual(sum_missing[2,1], 24)
        self.assertEqual(np.sum(sum_missing == 45), 7)

    def test_get_row_idx_with_highest_sum(self):
        sb = SudokuBoard()
        sb.place((5, 6), 3)
        self.assertEqual(sb.get_row_idx_with_highest_sum(), 5)

        to_insert = [((1, 0), 3), ((1, 7), 6), ((2, 1), 4), ((2, 2), 2), ((2, 8), 1), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_row_idx_with_highest_sum(), 1)

        to_insert = [((2, 3), 5), ((5, 8), 4), ((5, 1), 6)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_row_idx_with_highest_sum(), 5)

    def test_get_num_cols_where_number_is_present(self):
        sb = SudokuBoard()
        self.assertEqual(sb.get_num_cols_where_number_is_present(9), 0)
        self.assertEqual(sb.get_num_cols_where_number_is_present(1), 0)

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_num_cols_where_number_is_present(9), 0)
        self.assertEqual(sb.get_num_cols_where_number_is_present(3), 2)
        self.assertEqual(sb.get_num_cols_where_number_is_present(4), 3)
        self.assertEqual(sb.get_num_cols_where_number_is_present(8), 1)

    def test_get_num_empty_rows(self):
        sb = SudokuBoard()
        self.assertEqual(sb.get_num_empty_rows(), 9)
        sb.place((3, 3), 9)
        self.assertEqual(sb.get_num_empty_rows(), 8)

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_num_empty_rows(), 6)
        to_insert = [((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_num_empty_rows(), 4)

    def test_get_col_idxs_with_missing_num(self):
        sb = SudokuBoard()
        ret = sb.get_col_idxs_with_missing_num(1)
        self.assertEqual(ret.shape, (9,))
        self.assertEqual(tuple(np.sort(ret)), tuple(range(9)))

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_col_idxs_with_missing_num(1)
        self.assertEqual(ret.shape, (9,))
        self.assertEqual(tuple(np.sort(ret)), tuple(range(9)))
        ret = sb.get_col_idxs_with_missing_num(4)
        self.assertEqual(ret.shape, (6,))
        self.assertEqual(tuple(np.sort(ret)), (2,3,4,5,6,7))

        to_insert = [((4, 2), 4), ((1, 4), 4), ((3, 3), 4), ((0, 6), 4), ((6, 7), 4)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_col_idxs_with_missing_num(4)
        self.assertEqual(ret.shape, (1,))
        self.assertEqual(tuple(ret), (5,))

        sb.place((7, 5), 4)
        ret = sb.get_col_idxs_with_missing_num(4)
        self.assertEqual(ret.shape, (0,))

    def test_get_empty_block_idxs(self):
        sb = SudokuBoard()
        ret = sb.get_empty_block_idxs()
        self.assertEqual(ret.shape, (9, 2))
        self.assertEqual(tuple(np.sum(ret, axis=0)), (9, 9))

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_empty_block_idxs()
        self.assertEqual(ret.shape, (5, 2))
        self.assertEqual(tuple(np.sum(ret, axis=0)), (6, 5))

        to_insert = [((1, 4), 9), ((3, 2), 6), ((3, 3), 7), ((8, 5), 1), ((8, 8), 2)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_empty_block_idxs()
        self.assertEqual(ret.shape, (0, 2))

    def test_get_row_with_most_numbers(self):
        sb = SudokuBoard()
        sb.place((5,6), 3)
        most_numbers_row = sb.get_row_with_most_numbers()
        self.assertEqual(most_numbers_row.shape, (9,))
        self.assertTrue(np.all(most_numbers_row == [0,0,0,0,0,0,3,0,0]))

        to_insert = [((1,0), 3), ((1,7), 6), ((2,1), 4), ((3,7), 1)]
        [sb.place(*t) for t in to_insert]
        most_numbers_row = sb.get_row_with_most_numbers()
        self.assertEqual(most_numbers_row.shape, (9,))
        self.assertTrue(np.all(most_numbers_row == [3,0,0,0,0,0,0,6,0]))

        to_insert = [((2,3), 9), ((5,8), 4), ((5,1), 6)]
        [sb.place(*t) for t in to_insert]
        most_numbers_row = sb.get_row_with_most_numbers()
        self.assertEqual(most_numbers_row.shape, (9,))
        self.assertTrue(np.all(most_numbers_row == [0,6,0,0,0,0,3,0,4]))

    def test_get_valid_col_idxs_to_place(self):
        sb = SudokuBoard()
        self.assertEqual(len(sb.get_valid_col_idxs_to_place(5)), 9)

        sb.place((5,8), 5)
        self.assertEqual(len(sb.get_valid_col_idxs_to_place(5)), 8)

        sb.place((4,3), 6)
        self.assertEqual(len(sb.get_valid_col_idxs_to_place(5)), 8)

        to_insert = [((1,0), 5), ((2,7), 5), ((4,4), 5)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(len(sb.get_valid_col_idxs_to_place(5)), 5)

    def test_get_num_rows_with_two_nums_present(self):
        sb = SudokuBoard()
        self.assertEqual(sb.get_num_rows_with_two_nums_present(5, 7), 0)
        sb.place((3, 3), 9)
        self.assertEqual(sb.get_num_rows_with_two_nums_present(9, 1), 0)
        self.assertEqual(sb.get_num_rows_with_two_nums_present(9, 8), 0)
        sb.place((5, 3), 5)
        self.assertEqual(sb.get_num_rows_with_two_nums_present(9, 5), 0)

        to_insert = [((3, 8), 5), ((5, 2), 9), ((6, 2), 5), ((6, 3), 1)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_num_rows_with_two_nums_present(5, 9), 2)
        self.assertEqual(sb.get_num_rows_with_two_nums_present(9, 5), 2)
        self.assertEqual(sb.get_num_rows_with_two_nums_present(9, 1), 0)

        to_insert = [((6, 7), 9), ((7, 4), 9), ((0, 2), 2), ((0, 7), 5), ((5, 0), 3)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_num_rows_with_two_nums_present(5, 9), 3)
        self.assertEqual(sb.get_num_rows_with_two_nums_present(9, 3), 1)

    def test_get_max_of_rows_where_num_is_present(self):
        sb = SudokuBoard()
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(4), 0)
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(9), 0)

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(4), 4)
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(3), 4)
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(8), 8)
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(7), 0)

        to_insert = [((1, 4), 9), ((3, 2), 6), ((3, 3), 7), ((8, 5), 1), ((8, 8), 2)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(5), 0)
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(3), 9)
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(4), 4)
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(6), 8)
        self.assertEqual(sb.get_max_of_rows_where_num_is_present(1), 4)

    def test_get_rightmost_numbers_in_rows(self):
        sb = SudokuBoard()
        rightmost_numbers = sb.get_rightmost_numbers_in_rows()
        self.assertEqual(rightmost_numbers.shape, (9,))
        self.assertTrue(np.all(rightmost_numbers == np.zeros((9,))))

        sb.place((5, 6), 3)
        rightmost_numbers = sb.get_rightmost_numbers_in_rows()
        self.assertEqual(rightmost_numbers.shape, (9,))
        self.assertTrue(np.all(rightmost_numbers == [0,0,0,0,0,3,0,0,0]))

        to_insert = [((1, 0), 3), ((1, 7), 6), ((2, 1), 4), ((2, 2), 2), ((2, 8), 1), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        rightmost_numbers = sb.get_rightmost_numbers_in_rows()
        self.assertEqual(rightmost_numbers.shape, (9,))
        self.assertTrue(np.all(rightmost_numbers == [0,6,1,8,0,3,0,0,0]))

        to_insert = [((2, 3), 5), ((5, 8), 4), ((5, 1), 6), ((8, 0), 9), ((7, 8), 9)]
        [sb.place(*t) for t in to_insert]
        rightmost_numbers = sb.get_rightmost_numbers_in_rows()
        self.assertEqual(rightmost_numbers.shape, (9,))
        self.assertTrue(np.all(rightmost_numbers == [0,6,1,8,0,4,0,9,9]))

    def test_get_block_pos_with_minimal_max(self):
        sb = SudokuBoard()
        self.assertEqual(sb.get_block_pos_with_minimal_max(), None)
        sb.place((3, 3), 9)
        self.assertEqual(sb.get_block_pos_with_minimal_max(), (1, 1))

        to_insert = [((1, 0), 3), ((2, 7), 3)]
        [sb.place(*t) for t in to_insert]
        self.assertIn(sb.get_block_pos_with_minimal_max(), [(0, 0), (0, 2)])

        sb.place((2, 1), 4)
        self.assertEqual(sb.get_block_pos_with_minimal_max(), (0, 2))

        to_insert = [((5, 8), 1), ((8, 0), 2), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_block_pos_with_minimal_max(), (2, 0))

    def test_get_cols_with_above_average_sum(self):
        sb = SudokuBoard()
        ret = sb.get_cols_with_above_average_sum()
        self.assertTrue(ret.shape == (9, 0))

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_cols_with_above_average_sum()
        self.assertTrue(ret.shape == (9, 4))
        self.assertEqual(ret[1,0], 3)
        self.assertEqual(ret[2,0], 0)
        self.assertEqual(ret[2,1], 4)
        self.assertEqual(ret[2,2], 3)
        self.assertEqual(ret[3,2], 8)
        self.assertEqual(ret[5,3], 4)
        self.assertEqual(ret[5,1], 0)

        to_insert = [((1, 4), 9), ((3, 2), 6), ((3, 3), 7), ((8, 5), 1), ((8, 8), 2)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_cols_with_above_average_sum()
        self.assertTrue(ret.shape == (9, 6))
        self.assertEqual(ret[1,0], 3)
        self.assertEqual(ret[8,0], 4)
        self.assertEqual(ret[3,1], 6)
        self.assertEqual(ret[3,2], 7)
        self.assertEqual(ret[1,3], 9)
        self.assertEqual(ret[3,4], 8)
        self.assertEqual(ret[3,5], 0)
        self.assertEqual(ret[8,5], 2)

    def test_get_num_numbers_greater_than_col_avg(self):
        sb = SudokuBoard()
        self.assertEqual(sb.get_num_numbers_greater_than_col_avg(), 0)
        sb.place((3, 3), 9)
        self.assertEqual(sb.get_num_numbers_greater_than_col_avg(), 0)

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_num_numbers_greater_than_col_avg(), 2)
        to_insert = [((1, 4), 9), ((3, 2), 6), ((3, 3), 7), ((8, 5), 1), ((8, 8), 2), ((0, 8), 6)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_num_numbers_greater_than_col_avg(), 3)
        sb.place((1, 8), 7)
        self.assertEqual(sb.get_num_numbers_greater_than_col_avg(), 4)

    def test_get_num_numbers_smaller_than_row_avg(self):
        sb = SudokuBoard()
        self.assertEqual(sb.get_num_numbers_smaller_than_row_avg(), 0)
        sb.place((3, 3), 9)
        self.assertEqual(sb.get_num_numbers_smaller_than_row_avg(), 0)

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_num_numbers_smaller_than_row_avg(), 2)
        to_insert = [((1, 4), 9), ((3, 2), 6), ((3, 3), 7), ((8, 5), 1), ((8, 8), 2), ((0, 8), 6)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_num_numbers_smaller_than_row_avg(), 5)
        sb.place((1, 8), 7)
        self.assertEqual(sb.get_num_numbers_smaller_than_row_avg(), 5)

    def test_get_matching_nonempty_col_row_pair_idxs(self):
        sb = SudokuBoard()
        ret = sb.get_matching_nonempty_col_row_pair_idxs()
        self.assertTrue(ret.shape == (0, 2))

        to_insert = [((1, 1), 3), ((2, 7), 3), ((2, 0), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8), ((0, 5), 4), ((7, 5), 3)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_matching_nonempty_col_row_pair_idxs()
        self.assertTrue(ret.shape == (3, 2))
        self.assertTrue(np.all(np.unique(ret, axis=0) == np.array([[1, 1], [5, 2], [8, 0]])))   # unique sorts slices

        to_insert = [((7, 3), 9), ((0, 3), 5), ((6, 0), 5), ((6, 7), 9), ((8, 1), 2)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_matching_nonempty_col_row_pair_idxs()
        self.assertTrue(ret.shape == (2, 2))
        self.assertTrue(np.all(np.unique(ret, axis=0) == np.array([[3, 6], [5, 2]])))

    def test_get_number_frequencies(self):
        sb = SudokuBoard()
        self.assertEqual(sb.get_number_frequencies(1), 0)
        self.assertEqual(sb.get_number_frequencies(9), 0)

        to_insert = [((1, 0), 3), ((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_number_frequencies(1), 0)
        self.assertEqual(sb.get_number_frequencies(9), 0)
        self.assertEqual(sb.get_number_frequencies(4), 3)
        self.assertEqual(sb.get_number_frequencies(8), 1)
        self.assertEqual(sb.get_number_frequencies(4), 3)
        self.assertEqual(sb.get_number_frequencies(3), 2)

        to_insert = [((1, 4), 9), ((3, 2), 6), ((3, 3), 7), ((8, 5), 1), ((8, 8), 2)]
        [sb.place(*t) for t in to_insert]
        self.assertEqual(sb.get_number_frequencies(1), 1)
        self.assertEqual(sb.get_number_frequencies(9), 1)
        self.assertEqual(sb.get_number_frequencies(5), 0)
        self.assertEqual(sb.get_number_frequencies(4), 3)
        self.assertEqual(sb.get_number_frequencies(3), 2)

    def test_get_first_occurence_of_numbers(self):
        sb = SudokuBoard()
        ret = sb.get_first_occurence_of_numbers()
        self.assertTrue(len(ret) == 2)
        ix0, ix1 = ret
        self.assertTrue(np.all(ix0 == [9]*9) and np.all(ix1 == [9]*9))

        to_insert = [((1, 0), 3), ((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_first_occurence_of_numbers()
        self.assertTrue(len(ret) == 2)
        ix0, ix1 = ret
        self.assertTrue(np.all(ix0 == [9, 9, 1, 2, 9, 9, 9, 3, 9]) and np.all(ix1 == [9, 9, 0, 1, 9, 9, 9, 7, 9]))

        to_insert = [((1, 4), 9), ((3, 2), 6), ((3, 3), 7), ((8, 5), 1), ((8, 8), 2)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_first_occurence_of_numbers()
        self.assertTrue(len(ret) == 2)
        ix0, ix1 = ret
        self.assertTrue(np.all(ix0 == [8, 8, 1, 2, 9, 3, 3, 3, 1]) and np.all(ix1 == [5, 8, 0, 1, 9, 2, 3, 7, 4]))

    def test_get_second_largest_number_below_5_in_rows(self):
        sb = SudokuBoard()
        ret = sb.get_second_largest_number_below_5_in_rows()
        self.assertEqual(ret.shape, (9,))
        self.assertTrue(np.all(ret == [0,0,0,0,0,0,0,0,0]))

        sb.place((3, 3), 9)
        ret = sb.get_second_largest_number_below_5_in_rows()
        self.assertEqual(ret.shape, (9,))
        self.assertTrue(np.all(ret == [0,0,0,0,0,0,0,0,0]))

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_second_largest_number_below_5_in_rows()
        self.assertEqual(ret.shape, (9,))
        self.assertTrue(np.all(ret == [0,0,3,0,0,0,0,0,0]))

        to_insert = [((2, 8), 9), ((3, 0), 1), ((5, 1), 5), ((5, 2), 6), ((5, 3), 1)]
        [sb.place(*t) for t in to_insert]
        ret = sb.get_second_largest_number_below_5_in_rows()
        self.assertEqual(ret.shape, (9,))
        self.assertTrue(np.all(ret == [0,0,3,0,0,1,0,0,0]))

    def test_get_top5_num_with_pos(self):
        sb = SudokuBoard()
        top5_num_with_pos = sb.get_top5_num_with_pos()
        self.assertEqual(top5_num_with_pos.shape, (5,3))
        self.assertTrue(np.all(top5_num_with_pos == [[0,9,9]]))

        sb.place((3, 3), 9)
        top5_num_with_pos = sb.get_top5_num_with_pos()
        self.assertEqual(top5_num_with_pos.shape, (5,3))
        self.assertTrue(np.all(top5_num_with_pos == [[9,3,3],[0,9,9],[0,9,9],[0,9,9],[0,9,9]]))

        to_insert = [((1, 0), 3), ((2, 7), 3), ((2, 1), 4), ((5, 8), 4), ((8, 0), 4), ((3, 7), 8)]
        [sb.place(*t) for t in to_insert]
        top5_num_with_pos = sb.get_top5_num_with_pos()
        lexsorter = np.lexsort((top5_num_with_pos[:,2], top5_num_with_pos[:,1], top5_num_with_pos[:,0]))[::-1]
        top5_num_with_pos = top5_num_with_pos[lexsorter,:]

        self.assertEqual(top5_num_with_pos.shape, (5,3))
        self.assertTrue(np.all(top5_num_with_pos == [[9,3,3],[8,3,7],[4,8,0],[4,5,8],[4,2,1]]))

        to_insert = [((1, 4), 9), ((3, 2), 6), ((3, 3), 7), ((8, 5), 1), ((8, 8), 2), ((0, 8), 6)]
        [sb.place(*t) for t in to_insert]
        top5_num_with_pos = sb.get_top5_num_with_pos()
        lexsorter = np.lexsort((top5_num_with_pos[:,2], top5_num_with_pos[:,1], top5_num_with_pos[:,0]))[::-1]
        top5_num_with_pos = top5_num_with_pos[lexsorter,:]

        self.assertEqual(top5_num_with_pos.shape, (5,3))
        self.assertTrue(np.all(top5_num_with_pos == [[9,3,3],[9,1,4],[8,3,7],[6,3,2],[6,0,8]]))


    #

def suite():
    suite = unittest.TestSuite()
    testfuns = ["test_can_place", "test_place",
                "test_get_max_of_rows", "test_sum_of_missing_numbers_per_block",
                "test_get_row_idx_with_highest_sum", "test_get_num_cols_where_number_is_present",
                "test_get_num_empty_rows", "test_get_col_idxs_with_missing_num",
                "test_get_empty_block_idxs", "test_get_row_with_most_numbers",
                "test_get_valid_col_idxs_to_place", "test_get_num_rows_with_two_nums_present",
                "test_get_max_of_rows_where_num_is_present", "test_get_rightmost_numbers_in_rows",
                "test_get_block_pos_with_minimal_max", "test_get_cols_with_above_average_sum",
                "test_get_num_numbers_greater_than_col_avg", "test_get_num_numbers_smaller_than_row_avg",
                "test_get_matching_nonempty_col_row_pair_idxs",
                "test_get_number_frequencies", "test_get_first_occurence_of_numbers",
                "test_get_second_largest_number_below_5_in_rows", "test_get_top5_num_with_pos"
                ]
    [suite.addTest(TestSudoku(fun)) for fun in testfuns]
    return suite

runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite())
```
