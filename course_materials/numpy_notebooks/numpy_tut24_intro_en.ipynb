# **NumPy Array Programming Guide** - Introduction


In recent years, several areas of computer science have suddenly become popular where there is a need to process large amounts of data efficiently. Examples include various scientific computations and simulations, computer graphics, as well as data science and machine learning applications. What they have in common is that they apply large numbers of elementary mathematical and logical operations at the same time to large datasets. For this kind of massively and easily parallelizable computation, efficient dedicated hardware can be developed (so-called SIMD architectures: Single Instruction, Multiple Data). A good example of such hardware is the GPU, originally designed to support 3D computer graphics. GPUs often contain hundreds or thousands of simple execution units, and vectorized operations to be performed on large vectors or matrices can be distributed across them easily. Since deep neural networks, which are commonly used in machine learning today, are also built on this same type of computation, GPUs are extremely useful for training and evaluating neural networks as well.

To implement algorithms that operate on large amounts of data, specialized programming languages and techniques have also emerged. NVIDIA developed the CUDA programming interface for programming its general-purpose GPUs. This interface is very low-level: it provides direct access to GPU memory and execution units. It allows you to define algorithms that run individually on each thread/unit. However, this approach is inconvenient and overly verbose for implementing everyday, higher-level applications. A much higher-level solution is the **vectorized programming or array programming technique**). The main building blocks of array programming are arrays themselves, which, although they point to a contiguous region of memory, present the programmer with the abstraction of multidimensional tensors. With a special syntax, multiple elements of these seemingly multidimensional arrays can be referenced at once. As a result, complex data-processing operations can be defined in just a few lines using appropriate indexing and vector operations, often without using loops and branches.

Some of the characteristic syntactic elements of vectorized programming already appeared in a few early programming languages (for example, Fortran, or the fist purely array programming language, APL). The first truly popular language/software package supporting vectorized syntax was MATLAB. Nowadays, the NumPy package is one of the most fundamental extensions to Python, and a great many other Python packages build on it. Although NumPy itself only supports parallel execution to a limited extent and does not support using GPUs, the TensorFlow and PyTorch libraries — developed specifically for the efficient implementation of neural networks and supporting GPUs — also rely on NumPy’s syntax and implement most NumPy vector operations and indexing forms themselves.


## Briefly about the NumPy library


NumPy is one of the most important packages in Python: it is suitable for fast numerical computations. Python’s default compiler/interpreter is CPython. Compared to many other languages, this implementation produces programs that are orders of magnitude slower. The main reasons can be found in the dynamic typing and automatic memory management that make Python convenient, and in the fact that CPython functions almost entirely as an interpreter: the code’s prior analysis is largely limited to syntax checking, so there is no opportunity for code optimization.

Slow execution becomes most apparent when the interpreter has to execute a large number of Python statements—for example, when we implement matrix multiplication using loops and scalar multiplications. NumPy therefore provides functions that can perform the same operation in a vectorized way over large amounts of data, without having to use Python loops, for instance. NumPy functions call precompiled routines written in efficient languages (such as C or Fortran), so most of the computation runs using an efficient implementation, and Python is used only as the highest-level scripting language in our code.

Below is a short example where we generate two matrices filled with random numbers and then multiply them in the traditional way. We compare the efficiency of a naive Python implementation and a NumPy implementation.

First, let’s look at the naive, pure-Python code. For matrix multiplication we need to write three loops. It is clear that for larger matrices, millions of Python statements have to run. Our computer will therefore spend a lot of time executing code with the inefficient Python interpreter.


```python
# Pure Python implementation (matrices: lists of lists)

import random

def multiply_random_matrices_naive(matrix_dim):
  matrix1 = [[random.random() for col_idx in range(matrix_dim)] for row_idx in range(matrix_dim)]
  matrix2 = [[random.random() for col_idx in range(matrix_dim)] for row_idx in range(matrix_dim)]
  matrix_out = [[0. for col_idx in range(matrix_dim)] for row_idx in range(matrix_dim)]

  for row1_idx in range(matrix_dim):
    for col2_idx in range(matrix_dim):
      for item_idx in range(matrix_dim):
        matrix_out[row1_idx][col2_idx] += matrix1[row1_idx][item_idx] * matrix2[item_idx][col2_idx]

  return matrix_out
```

Next, let’s define the same thing using NumPy. We generate two matrices (arrays) filled with random elements using a total of two Python statements. We multiply these with matrix multiplication, which is a predefined operation in NumPy (`np.matmul()`). In total, only three Python statements will run, so the inefficient Python interpreter takes up only a minimal amount of time.


```python
# Numpy implementation (matrices: ndarrays)

import numpy as np

def multiply_random_matrices_npy(matrix_dim):
  matrix1 = np.random.rand(matrix_dim, matrix_dim)
  matrix2 = np.random.rand(matrix_dim, matrix_dim)

  return np.matmul(matrix1, matrix2)
```

Let’s measure the runtime of the two functions separately using a 400 × 400 matrix size.


```python
import time

# We measure the execution time

SIZE = 400
time_start = time.time()
multiply_random_matrices_naive(SIZE)
time_end = time.time()

print("Multiplying two ", SIZE, "x", SIZE, " random matrices in pure Python takes",\
          time_end - time_start, "seconds.")

time_start = time.time()
multiply_random_matrices_npy(SIZE)
time_end = time.time()

print("Multiplying two ", SIZE, "x", SIZE, " random matrices in Numpy takes",\
          time_end - time_start, "seconds.")
```

The naive matrix multiplication on 400 × 400 matrices took more than 16 seconds, while the NumPy implementation ran in around 20 milliseconds on Colab’s CPU. The NumPy implementation is therefore more than eight hundred times faster. Do not forget that both code snippets run on the CPU. With a GPU, an additional one to two orders of magnitude speedup can be achieved.


Besides efficiency, code conciseness can also be an argument in favor of NumPy. Consider solving a simple task in “bare-bones” Python versus an implementation using NumPy. Let’s count, in a list/array, how many elements are divisible by three or five:


```python
my_list = [2,4,5,6,3,9,11,1,15,0,22,18,17,5,8,4]

my_array = np.array(my_list)
```

Without NumPy we need to use a loop:


```python
c = 0
for elem in my_list:
  if (elem % 3 == 0) or (elem % 5 == 0):
    c += 1

print("Number of elements divisible by 3 or 5:", c)
```

Then, using NumPy:


```python
c = np.count_nonzero((my_array % 3 == 0) | (my_array % 5 == 0))

print("Number of elements divisible by 3 or 5:", c)
```

In almost every case where the operations performed in individual loop iterations are independent of each other, or where the loop implements a common programming pattern (for example, finding a maximum, counting, summation, collecting and counting certain elements, etc.), NumPy’s one-line statements can provide a much more concise solution.
