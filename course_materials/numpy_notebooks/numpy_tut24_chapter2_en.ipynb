## Referencing array slices (Basic indexing)



In the previous chapter we saw that NumPy allows us to perform operations over all elements of an array at once with a single statement. In many cases, however, we want to reference only a group of elements rather than all of them. We could do that with loops, but in Python this approach would be quite inefficient.

NumPy provides two different mechanisms for referencing a group of elements at once. In this chapter we discuss the first one, the so-called **basic indexing**, also known as slicing. Basic indexing lets us select intervals along the axes of an array (not necessarily with a step size of 1) and thereby reference multiple elements simultaneously.

It can be helpful to imagine a three-dimensional array as a rectangular cuboid. Within that cuboid we can select a contiguous region using axis-aligned cuts parallel to the faces. We can then reference all elements in the selected region. We can also choose different step sizes along each axis, in which case we reference only elements at the corresponding step positions within the region. The order of referenced elements can also be reversed independently along each axis by using negative step sizes.

Intervals with custom step sizes may already be familiar from the slicing syntax of `range()`, `np.arange()`, or standard Python lists. The syntax of basic indexing is similar, in fact identical to the list slicing notation (for example `my_list[2:5:-2]`), but for arrays we can specify a different interval (or a single index) for each axis.

First, let's create a one-dimensional array!


```python
import numpy as np

a = np.arange(10, dtype=np.int32)

print("The 'a' array:", a)
```

Consider different slices of the array `a`:


```python
print("a[2:7]:", a[2:7])
print("a[2:-2]:", a[2:-2])
print("a[-5:-2]:", a[-5:-2])
print("a[7:2]:", a[7:2])

print("\na[2:7:2]:", a[2:7:2])
print("a[2:7:-2]:", a[2:7:-2])
print("a[7:2:-2]:", a[7:2:-2])
```

Like for lists, the **syntax for slicing a one-dimensional array** is:

`arr[start_idx:end_idx_excl:step_size]`.

So we specify the start index, the (exclusive) stop index, and the step size, in that order. As before, **negative indices count backwards from the end of the axis**, so `-1` refers to the last element along the axis. The step size can also be negative; then we step backwards and the stop index must be smaller than the start index, otherwise no elements are selected.

The syntax can be shortened by omitting the second colon and the step size:

`arr[start_idx:end_idx_excl]`  $\ \ \ $- in that case the step size defaults to 1.

Let us look at a few of the examples above:
- `a[2:7]` yields `[2 3 4 5 6]`, because the exclusive stop index is `7`, so that element is not included.
- `a[2:-2]` yields `[2 3 4 5 6 7]`, because the array length is 10, so the stop index `-2` corresponds to 10-2 = 8.
- `a[7:2]` yields `[]`, because the start index is not smaller than the stop index and the step size is positive.
- Similarly, `a[2:7:-2]` also yields `[]`, because with a negative step the start index is smaller than the stop index.



```python
print("a[2:]:", a[2:])
print("a[:-2]:", a[:-2])

print("\na[2::2]:", a[2::2])
print("a[:-2:2]:", a[:-2:2])
print("a[::2]:", a[::2])
print("a[::-2]:", a[::-2])

print("\na[:]:", a[:])
print("a[::]:", a[::])
```

Any of the values can also be omitted:

`arr[start_idx:]` references elements from the given start index to the end of the axis, while `arr[:end_idx_excl]` references elements from the beginning up to the given exclusive stop index. `arr[:]` references all elements along the axis, and `arr[::step_size]` walks the full axis length with the (possibly negative) `step_size`.

An important characteristic of slice indexing is that **when using `:` as a slice, start or stop indices larger than the axis length do not raise an error**:


```python
print("The shape of 'a' is:", a.shape)

# print("a[12]:", a[12])    # IndexError: index 12 is out of bounds for axis 0 with size 10
print("a[12:]:", a[12:])    # no error
print("a[:12]:", a[:12])    # no error
```

For an array of shape (10,), `a[12]` is an invalid index, but `a[12:]` simply returns an empty array (because there is no valid index above 12 on that axis), and `a[:12]` returns the full array (because every valid index is smaller than 12).

The referenced elements can then be used for computation. In the example below we extract the differences between successive elements by applying element-wise subtraction between the slices `a[:-1]` and `a[1:]`:


```python
elems_all_but_last = a[:-1]
print("'a[:-1]' is ", elems_all_but_last)
print("The shape of 'a[:-1]' is:", elems_all_but_last.shape)

elems_all_but_first = a[1:]
print("\n'a[1:]' is ", elems_all_but_first)
print("The shape of 'a[1:]' is:", elems_all_but_first.shape)

diffs = elems_all_but_first - elems_all_but_last
print("\n'a[1:] - a[:-1]' is ", diffs)
```

Now let us examine what objects we get when we index a single element of an array, and what we get when we slice it using basic indexing:


```python
my_elem = a[4]

print("a[4]:", my_elem)
print("Type of a[4]:", type(my_elem))

my_slice = a[2:7]

print("\na[2:7]:", my_slice)
print("Type of a[2:7]:", type(my_slice))
```

Above you can see that `a[4]`, that is indexing a single element from the array, returns a scalar value whose type matches the array's dtype. In this case it is `np.int32`. This is similar to a Python `int`, but the latter (since Python 3) behaves like a BigInt/BigInteger type and can be arbitrarily large.

In contrast, indexing a slice (for example `a[2:7]`) returns an object of type `ndarray`, just like `a` itself. That is not surprising, because slicing a Python list also returns another list. The difference between list slicing and array slicing becomes clear in the next cell:


```python
# Python LIST

my_list = list(range(10))
print("The 'my_list' Python list:", my_list)

my_list_slice = my_list[2:7]
print("The 'my_list_slice' Python list:", my_list_slice)

my_list_slice[3] = 99   # writing into list slice

print("The 'my_list_slice' Python list after writing:", my_list_slice)
print("The 'my_list' Python list after writing 'my_list_slice':", my_list)

# NumPy ndarray

a = np.arange(10, dtype=np.int32)
print("\nThe 'a' array:", a)

my_slice = a[2:7]
print("The 'my_slice' array:", my_slice)

my_slice[3] = 99   # writing into array slice

print("The 'my_slice' array slice after writing into it:", my_slice)
print("The 'a' array after writing into 'my_slice':", a)
```

In the cell above we first create a standard Python list, then take a slice from it and assign it to a new variable. The slice is also a `list`. We then overwrite an element of the sliced list. The slice content changes, but the original list we sliced from remains unchanged. So Python list slicing performs a copy.

The situation is different for NumPy arrays. If we create an array and then take a slice from it using basic indexing, we obtain a new `ndarray`. If we overwrite an element of that new array, not only does the smaller sliced array change, but the original array changes as well. The reason is that **arrays obtained via slicing (basic indexing)** are not assigned to new memory regions. They are **views** of the original array. In short, the original array and all of its views change together.

The advantage of this approach is obvious: referencing a group of elements does not require allocating new memory. **When you slice an array using basic indexing, the resulting array points to a portion of the original array's memory.** Only the iterators used to enumerate elements (axis lengths via `ndarray.shape`, step sizes via `ndarray.strides`) differ compared to the original array.

Let us look at a few examples where we slice a multidimensional array:


```python
b = np.array([[0,1,2], [3,4,5]], dtype=np.int32)

print("The 'b' array:\n", b)

print("\nb[1:,:]:", b[1:,:])
print("b[:,1:]:\n", b[:,1:])
print("\nb[:-2,2:]:", b[:-2,2:])
print("b[1::-1,:2]:\n", b[1::-1,:2])
print("\nb[::-1,::-2]:\n", b[::-1,::-2])
```

In the cell above, for example, `b[1:,:]` selects the full matrix except the first row, while `b[:,1:]` selects everything except the first column. As you can see, in every case we used slice notation on both axes of the 2D array. This is not required, though. Slicing occurs even if we replace _some_ (but not all!) of the slice indices with single-element indices:

```python
my_elem = b[1,1]

print("b[1,1]:", my_elem)
print("Type of b[1,1]:", type(my_elem))

my_slice1 = b[1:,1]

print("\nb[1:,1]:", my_slice1)
print("Type of b[1:,1]:", type(my_slice1))

my_slice2 = b[1,::2]

print("\nb[1,::2]:", my_slice2)
print("Type of b[1,::2]:", type(my_slice2))

my_slice3 = b[:,1]

print("\nb[:,1]:", my_slice3)
print("Type of b[:,1]:", type(my_slice3))
```

In the cell above only the very first example does not produce slicing, that is basic indexing. `b[1,1]` specifies a single index along both axes, so it necessarily returns a single element: the entry in the second row and second column. The type of the returned value will be an `np.int32` scalar.

In contrast, `b[:,1]` references the full second column, and `b[1,::2]` references every second element of the second row. These do not necessarily return a single element, so the referenced values are placed into a new array view.

It is worth comparing the `a[2,:]` and `a[2:3, :]` cases. Both expressions reference the same elements, but along the first (axis #0) dimension the former uses a single index, so that axis disappears in the result view and the view becomes one-dimensional (with shape `(5,)`). In the latter expression both axes are sliced, so both axes remain in the result view (with shape `(1, 5)`).

As shown below, if you specify a slice or a single index only for the first few axes and you want to select all elements (`:`) along the remaining axes, you can omit those trailing `:` entries:


```python
print("b[1,:]:", b[1,:])

print("b[1]:", b[1])
```

Conversely, if you want to specify indices only for the last few axes, you can use the `...` (ellipsis) index:


```python
print("b[:,0]:", b[:,0])

print("b[...,0]:", b[...,0])
```

Slice indexing can also be used on the left-hand side of an assignment, so you can overwrite elements inside array slices. In the assignment `b[1,:] = 42` shown below, we are writing into a view of `b`. Keep in mind that if you write into a view, the original array and its other views also change:


```python
b = np.array([[0,1,2], [3,4,5]], dtype=np.int32)

print("The 'b' array:\n", b)

b[1,:] = 42
b[0,::2] = 99

print("\nThe 'b' array after writing into it:\n", b)
```

## Changing the shape of arrays


As mentioned at the beginning of the previous chapter, the N-dimensional shape of `ndarray` arrays is only apparent. In memory, the elements are stored sequentially for any shape (as long as we are not dealing with a view of another array).

So there is nothing preventing us from taking a 6-element array that currently has the one-dimensional shape (6,) and reshaping it into a two-dimensional array of shape (3,2). The word "reshape" might even be misleading here, because **in this case as well we only create a new two-dimensional view over the original array**.

New views with a different shape can be created using `ndarray.reshape()` or the equivalent `np.reshape()`. You must provide the new shape:


```python
b = np.arange(6, dtype=np.int32)

print("The shape of the 'b' array:", b.shape)
print("The 'b' array:", b)

b2 = b.reshape(3,2)

print("\nThe shape of the 'b2' array (view):", b2.shape)
print("The 'b2' array (view):\n", b2)

b3 = b2.reshape(6)

print("\nThe shape of the 'b3' array (view):", b3.shape)
print("The 'b3' array (view):\n", b3)
```

As we already observed for array slices, writing into a view changes the content of the original array and its other views as well.


```python
b[4] = 42
b2[2,1] = 99
b3[1:3] = 77

print("The 'b' array:", b)
print("The 'b2' array (view):\n", b2)
print("\nThe 'b3' array (view):", b3)
```

As the number of elements in an array is fixed, the product of the axis lengths in the shape is also fixed. When reshaping, one axis length can therefore be inferred from the others. **If you replace one axis length with `-1`, that length is computed automatically**. For example, `arr.reshape(-1)` returns a one-dimensional (flattened / flat) view of `arr`.


```python
print("The shape of the 'b2' array (view):", b2.shape)
print("The 'b2' array (view):\n", b2)

b4 = b2.reshape(-1)   # equivalent to: b4 = np.reshape(b2, (-1,))

print("\nThe shape of the 'b4' array (view):", b4.shape)
print("The 'b4' array (view):", b4)

b5 = b2.reshape(-1, 1)

print("\nThe shape of the 'b5' array (view):", b5.shape)
print("The 'b5' array (view):\n", b5)

b6 = b2.reshape(1, 1, 2, -1, 1)

print("\nThe shape of the 'b6' array (view):", b6.shape)
print("The 'b6' array (view):\n", b6)
```

**By using `None` or the equivalent `np.newaxis` as an index, you can create new axes of length 1 in the shape**. This convenience will become important for shape matching (broadcasting), see [Chapter 3](https://colab.research.google.com/drive/1MHHB8hkZZs9OhUep024x9anZrZ9L5SAv).



```python
b = np.arange(6, dtype=np.int32)

print("The shape of the 'b' array:", b.shape)
print("The 'b' array:", b)

print("\nb[None,:].shape:", b[None,:].shape)
print("b[:,None].shape:", b[:,np.newaxis].shape)     # None is equivalent to np.newaxis in this role
```

In the cell above you can see that for the one-dimensional array `b` with shape (6,), the expression `b[None,:]` creates a view with shape (1,6), while `b[:,None]` creates a view with shape (6,1). The former is equivalent to `b.reshape(1, -1)` and the latter to `b.reshape(-1, 1)`. In this role, the keyword `None` and the symbol `np.newaxis` are equivalent.

`None`/`np.newaxis` indexing can be freely combined with other kinds of indexing:


```python
print("b[None,:,None,None].shape:", b[None,:,None,None].shape)
print("b[np.newaxis,:,np.newaxis,np.newaxis].shape:", b[np.newaxis, :, np.newaxis, np.newaxis].shape)  # same as line above

b_2d = b.reshape(3,2)

print("\nThe shape of the 'b_2d' array (view):", b_2d.shape)
print("The 'b_2d' array (view):\n", b_2d)

print("\nb_2d[None,:,None,None,:].shape:", b_2d[None,:,None,None,:].shape)
print("b_2d[1::2,None,0].shape:", b_2d[1::2,None,0].shape)
print("b_2d[None,::-1,...,None].shape:", b_2d[None,::-1,...,None].shape)
```

## `np.argmax()`, `np.argmin()` operations


`np.argmax()` and `np.argmin()` can also be categorized as the *vector $\rightarrow$ scalar* (collapse) operations discussed at the end of Chapter 1, but to understand their output it helps to know the concept of a flattened (one-dimensional, flat) array.

On a simple one-dimensional array, `np.argmax()` returns the index of the largest element, while `np.argmin()` returns the index of the smallest element. If there are multiple elements with the same extreme value, the first such index is returned:


```python
arr = np.array([4,1,5,1,7], dtype=np.int32)
print("The 1D 'arr' array:", arr)

print("\nnp.argmax(arr):", np.argmax(arr))
print("np.argmin(arr):", np.argmin(arr))
```

When applied to a multidimensional array, however, these functions do not return indices along each axis. Instead, they return the index in the flattened (flat) array. In the code below, although the maximum of the (2, 3) array is at index `[1,2]`, `np.argmax()` returns 5, because in row-major order the element at `[1,2]` is the sixth element (0-based index 5). If we flatten the 2D array (create a 1D view), the index `[5]` then correctly refers to the maximal element:


```python
arr2d = np.arange(6, dtype=np.int32).reshape(2,3) * 10

print("The 2D 'arr2d' array:\n", arr2d)
print("   ... its shape is", arr2d.shape)

max_flat_idx = np.argmax(arr2d)
print("\nnp.argmax(arr2d):", max_flat_idx)

# print("arr2d[max_flat_idx]:", arr2d[max_flat_idx])   # IndexError, idx out of bounds for axis#0 - code would only work with 1D array

arr2d_flat = arr2d.reshape(-1)
print("\narr2d.reshape(-1)[max_flat_idx]:", arr2d_flat[max_flat_idx])
```

`np.argmin()` and `np.argmax()` have a special property: unlike most NumPy functions, they can exhibit *short-circuit* behavior, meaning that in certain cases they can stop scanning the input array early. If, while searching for the minimum/maximum, they encounter a value equal to the dtype's minimum/maximum, they stop and return that element's index.

This behavior can be used to efficiently (in linear time) find the first occurrence of certain values in arrays. For example, if the dtype of the argument array is `np.bool_`, then `np.argmax()` stops as soon as it reaches the first `True` element and returns its index. `np.argmax(arr == 42)` for instance returns the index of the first element equal to `42` in `arr`. In contrast, `np.amax()` (and most functions in vectorized languages) cannot stop scanning early.


## Conversion between multidimensional and flattened indices


From the above we learned that when `np.argmin()` and `np.argmax()` are applied to multidimensional arrays without an `axis` parameter, they return the index of the smallest and largest element, respectively. In that case the index does not index the individual axes of the multidimensional array, but its flattened, one-dimensional (flat) version.

In the cells above we used this flat index by first creating a 1D view via `ndarray.reshape(-1)`, and then indexing that view with the flat index. However, there is another way to use the returned index: we do not necessarily need to create a flattened view.

`np.unravel_index()` converts an index that refers to the flattened view into an index (or indices) that refer to another, possibly multidimensional, view of the same array:


```python
arr2d = np.arange(6, dtype=np.int32).reshape(2,3) * 10

print("The 2D 'arr2d' array:\n", arr2d)
print("   ... its shape is", arr2d.shape)

max_flat_idx = np.argmax(arr2d)
print("\nnp.argmax(arr2d):", max_flat_idx)

# indexing the flat (1D) view of the 2D array with the flat index
arr2d_flat = arr2d.reshape(-1)
print("\narr2d.reshape(-1)[max_flat_idx]:", arr2d_flat[max_flat_idx])

# converting the flat index to a multidimensional index
max_2d_idx = np.unravel_index(max_flat_idx, shape=arr2d.shape)
print("\nnp.unravel_index(max_flat_idx, shape=arr2d.shape):", max_2d_idx)

# indexing the 2D array directly with the multidimensional index
print("\narr2d[max_2d_idx]:", arr2d[max_2d_idx])
```

The inverse conversion, that is "flattening" multidimensional indices into indices for a one-dimensional view, can be done with `np.ravel_multi_index()`. Both of the mentioned functions can convert multiple indices at once, so you can also pass arrays of indices.

The resulting index arrays can be used to reference multiple elements simultaneously using the *advanced indexing* syntax (see [Chapter 4](numpy_tut24_chapter4_en.ipynb)). If `np.unravel_index()` receives multiple flat indices, it returns the multidimensional indices in the same format as the output of `np.where()` (see [Chapter 5](numpy_tut24_chapter5_en.ipynb)).
