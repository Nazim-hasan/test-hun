## **Gradebook**

Create a class that stores students’ names, their Neptun codes, and their grades. The Neptun code uniquely identifies students. The class should be named **Gradebook**. Its public methods must be the following:

* **add_student(student_name, student_neptun_code)**: Store the given student’s name and Neptun code in the gradebook if there is no student with the same code yet. If the Neptun code is already present in the gradebook, raise `StudentAlreadyExistsException`, and include the code in the exception message.
* **add_students(student_list)**: Store the given students’ names and Neptun codes in the gradebook if none of their codes are already present, and if there are no duplicate Neptun codes among the students to be added. `student_list` is a list of 2-element tuple objects; the two elements are, in order, the student’s name and Neptun code. If any Neptun code is already in the gradebook, or if the same Neptun code occurs more than once in `student_list`, the function must raise `StudentAlreadyExistsException`, and include one of the duplicated codes in the exception message.
* **drop_student(student_neptun_code)**: Remove the student (and their grades) with the given Neptun code from the gradebook, if such a student exists. If no student with that code exists, raise `StudentMissingException`, and include the code in the exception message.
* **get_student_name(student_neptun_code)**: Return the name of the student with the given Neptun code. If no student with that Neptun code exists in the gradebook, raise `StudentMissingException`, and include the missing Neptun code in the exception message.
* **get_all_student_neptun_codes()**: Return the Neptun codes stored in the gradebook as a list, sorted alphabetically.
* **get_all_student_names()**: Return the student names stored in the gradebook as a list, sorted alphabetically.
* **search_students_by_name(student_namepart)**: Return a list of students whose name contains the string `student_namepart` as a substring. The function returns a list of 2-element tuple objects; the tuple elements are, in order, the student’s name and Neptun code. If there is no matching student, return an empty list.
* **add_grade(student_neptun_code, grade)**: Add the given grade to the student with the given Neptun code. If no student with that code exists in the gradebook, raise `StudentMissingException`, and include the missing code in the exception message.
* **get_student_grades(student_neptun_code)**: Return the grades of the student with the given Neptun code as a list. If no student with that code exists in the gradebook, raise `StudentMissingException`, and include the missing code in the exception message.
* **get_number_of_students_with_low_grades()**: Return how many students have at least one grade of 1 or 2.
* **get_average_grade()**: Return the average of all grades present in the gradebook. If there are no grades in the gradebook, return `0`.
* **get_avarage_grade_per_student()**: Return each student’s average grade in a dictionary (`dict`). The keys are 2-element tuple objects containing, in order, the student’s name and Neptun code; the values are the students’ averages as numbers. If a student has no grades recorded, return `0` for that student.
* **get_worst_recent_grade()**: Return, for each student who has grades, the minimum of their last five grades in a dictionary (`dict`). The keys are 2-element tuple objects containing, in order, the student’s name and Neptun code; the values are the minima of the last five grades as numbers. If a student has no grades recorded, they must not appear in the returned dictionary.
* **get_student_codes_with_many_5s()**: Return a list containing the codes of those students who have more grades of 5 than all their other grades combined.


### Solution


```python

class StudentAlreadyExistsException(Exception):
    pass

class StudentMissingException(Exception):
    pass



class Gradebook:

    def __init__(self):
        self.grades = {}

    # self.grades =
    # {
    #    'neptun_code1': ('Student Name1', [4, 5, 2, 4]),
    #    'neptun_code2': ('Student Name2', [3, 1, 2, 5, 5]),
    #  }

    def add_student(self, student_name, student_neptun_code):
        if student_neptun_code in self.grades:
            raise StudentAlreadyExistsException(f"Student already in gradebook with neptun code: {student_neptun_code}")

        self.grades[student_neptun_code] = (student_name, [])

    def add_students(self, student_list):
        # student_list: [(student_name, student_neptun_code)]

        # Veryfing that there are no duplicates among the to be added students and none of them are already in the gradebook
        neptun_codes_to_add = set()
        for idx in range(len(student_list)):
            student_neptun_code = student_list[idx][1]
            if (student_neptun_code in self.grades) or (student_neptun_code in neptun_codes_to_add):
                raise StudentAlreadyExistsException(f"Student already in gradebook with neptun code: {student_neptun_code}")
            neptun_codes_to_add.add(student_neptun_code)

        # Adding students to the gradebook
        for idx in range(len(student_list)):
            student_name, student_neptun_code = student_list[idx]
            self.grades[student_neptun_code] = (student_name, [])

        # An alternative solution: 
        #     (deep)copy the dictionary
        #     try adding students one by one with self.add_student(..)
        #         -> if we get any exception, reset self.grades to the copy with the original state

    def drop_student(self, student_neptun_code):
        if student_neptun_code not in self.grades:
            raise StudentMissingException(f"Student not present in gradebook with neptun code: {student_neptun_code}")

        del self.grades[student_neptun_code]

    # Alternative approach
    def drop_student(self, student_neptun_code):
        try:
            self.grades.pop(student_neptun_code)      # An alternative to del self.grades[student_neptun_code]
        except KeyError:
            raise StudentMissingException(f"Student not present in gradebook with neptun code: {student_neptun_code}")

    def get_student_name(self, student_neptun_code):
        try:
            return self.grades[student_neptun_code][0]
        except KeyError:
            raise StudentMissingException(f"Student not present in gradebook with neptun code: {student_neptun_code}")

    def get_all_student_neptun_codes(self):
        return sorted(list(self.grades.keys()))

    def get_all_student_names(self):
        list_of_all_student_names = []
        for name, _ in self.grades.values():
            list_of_all_student_names.append(name)

        return sorted(list_of_all_student_names)

    # Alternative approach
    def get_all_student_names(self):
    
        def func_get_first_elem(seq):
            return seq[0]

        return sorted(map(func_get_first_elem, self.grades.values()))

    # Alternative approach #2
    def get_all_student_names(self):
        return sorted(map(lambda seq: seq[0], self.grades.values()))
    
    def search_students_by_name(self, student_namepart):
        found_students = []
        for neptun_code, (name, _) in self.grades.items():
            if student_namepart in name:
                found_students.append((name, neptun_code))
                
        return found_students

    def add_grade(self, student_neptun_code, grade):
        if student_neptun_code in self.grades:
            self.grades[student_neptun_code][1].append(grade)
        else:
            raise StudentMissingException(f"Student not present in gradebook with neptun code: {student_neptun_code}")

    def get_student_grades(self, student_neptun_code):
        if student_neptun_code in self.grades:
            return self.grades[student_neptun_code][1]
        else:
            raise StudentMissingException(f"Student not present in gradebook with neptun code: {student_neptun_code}")

    def get_number_of_students_with_low_grades(self):
        c = 0
        for _, student_grades_list in self.grades.values():
            if (1 in student_grades_list) or (2 in student_grades_list):
                c += 1
        return c
        
    # Alternative approach
    def get_number_of_students_with_low_grades(self):
        c = 0
        for _, student_grades_list in self.grades.values():
            if len(set(range(1,3)) & set(student_grades_list)) > 0:      # An alternative to: if len({1,2} & set(student_grades_list)) > 0:
                c += 1
        return c

    def get_average_grade(self):
        grades_sum, grades_count = 0, 0
        for _, student_grades_list in self.grades.values():
            grades_sum += sum(student_grades_list)
            grades_count += len(student_grades_list)
        return 0 if grades_count == 0 else grades_sum / grades_count     # an inline version of the if statement


    def get_avarage_grade_per_student(self):
        avg_grades_dict = {}
        for neptun_code, (name, student_grades_list) in self.grades.items():
            avg_grade = 0 if len(student_grades_list) == 0 else sum(student_grades_list) / len(student_grades_list)
            avg_grades_dict[(name, neptun_code)] = avg_grade
        return avg_grades_dict
        
    def get_worst_recent_grade(self):
        worst_grades_dict = {}
        for neptun_code, (name, student_grades_list) in self.grades.items():
            if len(student_grades_list) > 0:
                worst_grades_dict[(name, neptun_code)] = min(student_grades_list[-5:])
        return worst_grades_dict
        
    def get_student_codes_with_many_5s(self):
        good_grade_students = []
        for neptun_code, (_, student_grades_list) in self.grades.items():
            count5s = student_grades_list.count(5)
            if count5s > len(student_grades_list) - count5s:
                good_grade_students.append(neptun_code)
        return good_grade_students
        
# ----- Solutions with comprehensions -----

    '''
    def add_students(self, student_list):
        # student_list: [(student_name, student_neptun_code)]
        # Duplication checking should be added here: use the classic solution perhaps
        students_dict_to_add = {code: (name, []) for name, code in student_list}
        self.grades.update(students_dict_to_add)
    '''

    def get_all_student_names(self):
        return sorted([name for name, _ in self.grades.values()])

    def search_students_by_name(self, student_namepart):
        return [(name, neptun_code) for neptun_code, (name, _) in self.grades.items() if student_namepart in name]

    def get_number_of_students_with_low_grades(self):
        #return len([_ for _, st_grades in self.grades.values() if (1 in st_grades) or (2 in st_grades)])
        return len([_ for _, st_grades in self.grades.values() if len(set(range(1,3)) & set(st_grades)) > 0])

    def get_average_grade(self):
        grades_sum = sum([sum(st_grades) for _, st_grades in self.grades.values()])
        grades_count = sum([len(st_grades) for _, st_grades in self.grades.values()])
        return 0 if grades_count == 0 else grades_sum / grades_count

    def get_avarage_grade_per_student(self):
        return {(name, code): 0 if len(st_grades) == 0 else sum(st_grades) / len(st_grades) \
                                            for code, (name, st_grades) in self.grades.items()}

    def get_worst_recent_grade(self):
        return {(name, code): min(st_grades[-5:]) \
                            for code, (name, st_grades) in self.grades.items() if len(st_grades) > 0}

    def get_student_codes_with_many_5s(self):
        return [code for code, (_, st_grades) in self.grades.items() \
                            # if st_grades.count(5) > len(st_grades) - st_grades.count(5)]     # inefficient because of counting the same thing twice
                            if (len(st_grades) > 0) and (st_grades.count(5) / len(st_grades) > 0.5)]

    def get_student_codes_with_many_5s(self):
        return [code for code, (_, st_grades) in self.grades.items() \
                            if (c_grades := st_grades.count(5)) > len(st_grades) - c_grades]

# ----- Solutions with generator expressions -----

    def get_all_student_names(self):
        return sorted((name for name, _ in self.grades.values()))

    def get_average_grade(self):
        grades_sum = sum((sum(st_grades) for _, st_grades in self.grades.values()))
        grades_count = sum((len(st_grades) for _, st_grades in self.grades.values()))
        return 0 if grades_count == 0 else grades_sum / grades_count

        
        
```

### Tests


```python
import unittest

class TestGradebook(unittest.TestCase):

    def test_add_student(self):
        # only tests whether exception is raised
        gb = Gradebook()
        gb.add_student("Ariona Pirku", "abcd12")
        gb.add_student("Ariona Pirku", "efg12h")
        self.assertRaises(StudentAlreadyExistsException, gb.add_student, "Bernadett Fegyó", "efg12h")

    def test_add_students(self):
        # only tests whether exception is raised
        gb = Gradebook()
        students1 = [("Arpita Ekka", "abcd12"), ("Arpita Ekka", "efg12h")]
        gb.add_students(students1)
        self.assertRaises(StudentAlreadyExistsException, gb.add_students, students1)
        students2 = [("Bernadett Fegyó", "h56ijk"), ("Arpita Ekka", "87mnop")]
        gb.add_students(students2)
        students3 = [("Bernadett Fegyó", "st9uv0"), ("Emmett Taylor Elliott", "st9uv0")]
        self.assertRaises(StudentAlreadyExistsException, gb.add_students, students3)

    def test_drop_student(self):
        gb = Gradebook()
        gb.add_student("Ariona Pirku", "abcd12")
        gb.add_student("Ariona Pirku", "efg12h")
        self.assertRaises(StudentMissingException, gb.drop_student, "87mnop")
        gb.drop_student("abcd12")
        self.assertRaises(StudentMissingException, gb.drop_student, "abcd12")
        gb.drop_student("efg12h")
        gb.add_student("Ariona Pirku", "efg12h")
        gb.drop_student("efg12h")

        gb = Gradebook()
        students1 = [("Ariona Pirku", "abcd12"), ("Ariona Pirku", "efg12h")]
        gb.add_students(students1)
        self.assertRaises(StudentMissingException, gb.drop_student, "87mnop")
        gb.drop_student("abcd12")
        self.assertRaises(StudentMissingException, gb.drop_student, "abcd12")
        gb.drop_student("efg12h")

    def test_get_student_name(self):
        gb = Gradebook()
        self.assertRaises(StudentMissingException, gb.get_student_name, "87mnop")
        students1 = [("Emmett Taylor Elliott", "abcd12"), ("Tamara Takács", "efg12h")]
        gb.add_students(students1)
        self.assertEqual("Emmett Taylor Elliott", gb.get_student_name("abcd12"))
        gb.drop_student("abcd12")
        self.assertRaises(StudentMissingException, gb.get_student_name, "abcd12")
        self.assertEqual("Tamara Takács", gb.get_student_name("efg12h"))
        gb.drop_student("efg12h")
        self.assertRaises(StudentMissingException, gb.get_student_name, "efg12h")

    def test_get_all_student_neptun_codes(self):
        gb = Gradebook()
        self.assertEqual([], gb.get_all_student_neptun_codes())
        students1 = [("Arpita Ekka", "efg12h"), ("Arpita Ekka", "abcd12")]
        gb.add_students(students1)
        self.assertEqual(["abcd12", "efg12h"], gb.get_all_student_neptun_codes())
        gb.drop_student("efg12h")
        self.assertEqual(["abcd12"], gb.get_all_student_neptun_codes())
        gb.drop_student("abcd12")
        self.assertEqual([], gb.get_all_student_neptun_codes())
        gb.add_student("Arpita Ekka", "efg12h")
        self.assertEqual(["efg12h"], gb.get_all_student_neptun_codes())

    def test_get_all_student_names(self):
        gb = Gradebook()
        self.assertEqual([], gb.get_all_student_names())
        students1 = [("Bernadett Fegyó", "efg12h"), ("Bernadett Fegyó", "abcd12")]
        gb.add_students(students1)
        self.assertEqual(["Bernadett Fegyó", "Bernadett Fegyó"], gb.get_all_student_names())
        gb.drop_student("efg12h")
        self.assertEqual(["Bernadett Fegyó"], gb.get_all_student_names())
        gb.drop_student("abcd12")
        self.assertEqual([], gb.get_all_student_names())
        gb.add_student("Bernadett Fegyó", "efg12h")
        gb.add_student("Emmett Taylor Elliott", "abcd12")
        self.assertEqual(["Bernadett Fegyó", "Emmett Taylor Elliott"], gb.get_all_student_names())

    def test_search_students_by_name(self):
        gb = Gradebook()
        self.assertEqual([], gb.search_students_by_name("kk"))
        self.assertEqual([], gb.search_students_by_name(" "))
        students1 = [("Arpita Ekka", "efg12h"), ("Arpita Ekka", "abcd12")]
        gb.add_students(students1)
        ret1 = gb.search_students_by_name("kk")
        self.assertIs(type(ret1), list)
        self.assertEqual([("Arpita Ekka", "abcd12"), ("Arpita Ekka", "efg12h")], sorted(ret1))
        students2 = [("Ariona Pirku", "h56ijk"), ("Tamara Takács", "87mnop")]
        gb.add_students(students2)
        self.assertEqual([], gb.search_students_by_name("Margit"))
        self.assertEqual([("Arpita Ekka", "abcd12"), ("Arpita Ekka", "efg12h")], sorted(gb.search_students_by_name("rpi")))
        self.assertEqual([("Ariona Pirku", "h56ijk"), ("Arpita Ekka", "abcd12"), \
                    ("Arpita Ekka", "efg12h"), ("Tamara Takács", "87mnop")], sorted(gb.search_students_by_name(" ")))

    def test_add_grade(self):
        # only tests whether exception is raised
        gb = Gradebook()
        self.assertRaises(StudentMissingException, gb.add_grade, "efg12h", 3)
        students1 = [("Ariona Pirku", "efg12h"), ("Ariona Pirku", "abcd12")]
        gb.add_students(students1)
        gb.add_grade("efg12h", 3)
        gb.add_grade("efg12h", 3)
        gb.add_grade("abcd12", 1)
        self.assertRaises(StudentMissingException, gb.add_grade, "87mnop", 3)

    def test_get_student_grades(self):
        gb = Gradebook()
        self.assertRaises(StudentMissingException, gb.get_student_grades, "efg12h")
        students1 = [("Emmett Taylor Elliott", "efg12h"), ("Emmett Taylor Elliott", "abcd12")]
        gb.add_students(students1)
        self.assertRaises(StudentMissingException, gb.get_student_grades, "87mnop")
        gb.add_grade("efg12h", 3)
        gb.add_grade("efg12h", 4)
        self.assertEqual([], gb.get_student_grades("abcd12"))
        gb.add_grade("abcd12", 1)
        self.assertEqual([1], gb.get_student_grades("abcd12"))
        self.assertEqual([3, 4], gb.get_student_grades("efg12h"))
        gb.drop_student("abcd12")
        self.assertRaises(StudentMissingException, gb.get_student_grades, "abcd12")

    def test_get_number_of_students_with_low_grades(self):
        gb = Gradebook()
        self.assertEqual(0, gb.get_number_of_students_with_low_grades())
        students1 = [("Emmett Taylor Elliott", "efg12h"), ("Emmett Taylor Elliott", "abcd12")]
        gb.add_students(students1)
        self.assertEqual(0, gb.get_number_of_students_with_low_grades())
        gb.add_grade("efg12h", 3)
        gb.add_grade("efg12h", 4)
        self.assertEqual(0, gb.get_number_of_students_with_low_grades())
        gb.add_grade("abcd12", 1)
        self.assertEqual(1, gb.get_number_of_students_with_low_grades())
        gb.add_grade("efg12h", 2)
        self.assertEqual(2, gb.get_number_of_students_with_low_grades())
        gb.drop_student("abcd12")
        self.assertEqual(1, gb.get_number_of_students_with_low_grades())

    def test_get_average_grade(self):
        gb = Gradebook()
        self.assertEqual(0, gb.get_average_grade())
        students1 = [("Bernadett Fegyó", "efg12h"), ("Bernadett Fegyó", "abcd12")]
        gb.add_students(students1)
        self.assertEqual(0, gb.get_average_grade())
        gb.add_grade("efg12h", 3)
        gb.add_grade("efg12h", 4)
        self.assertAlmostEqual(3.5, gb.get_average_grade())
        gb.add_grade("abcd12", 1)
        self.assertAlmostEqual(8/3, gb.get_average_grade())
        gb.add_grade("efg12h", 2)
        self.assertAlmostEqual(2.5, gb.get_average_grade())
        gb.drop_student("abcd12")
        self.assertAlmostEqual(3, gb.get_average_grade())

    def test_get_avarage_grade_per_student(self):
        gb = Gradebook()
        self.assertEqual({}, gb.get_avarage_grade_per_student())
        students1 = [("Bernadett Fegyó", "efg12h"), ("Tamara Takács", "abcd12")]
        gb.add_students(students1)
        self.assertEqual({("Bernadett Fegyó", "efg12h"): 0, ("Tamara Takács", "abcd12"): 0}, gb.get_avarage_grade_per_student())
        gb.add_grade("efg12h", 2)
        gb.add_grade("efg12h", 4)
        self.assertEqual({("Bernadett Fegyó", "efg12h"): 3, ("Tamara Takács", "abcd12"): 0}, gb.get_avarage_grade_per_student())
        gb.add_grade("abcd12", 1)
        self.assertEqual({("Bernadett Fegyó", "efg12h"): 3, ("Tamara Takács", "abcd12"): 1}, gb.get_avarage_grade_per_student())
        gb.drop_student("abcd12")
        self.assertEqual({("Bernadett Fegyó", "efg12h"): 3}, gb.get_avarage_grade_per_student())

    def test_get_worst_recent_grade(self):
        gb = Gradebook()
        self.assertEqual({}, gb.get_worst_recent_grade())
        students1 = [("Arpita Ekka", "efg12h"), ("Tamara Takács", "abcd12")]
        gb.add_students(students1)
        self.assertEqual({}, gb.get_worst_recent_grade())
        gb.add_grade("efg12h", 2)
        gb.add_grade("efg12h", 4)
        self.assertEqual({("Arpita Ekka", "efg12h"): 2}, gb.get_worst_recent_grade())
        gb.add_grade("abcd12", 1)
        self.assertEqual({("Arpita Ekka", "efg12h"): 2, ("Tamara Takács", "abcd12"): 1}, gb.get_worst_recent_grade())
        gb.add_grade("efg12h", 3)
        gb.add_grade("efg12h", 4)
        gb.add_grade("efg12h", 5)
        gb.add_grade("efg12h", 5)
        gb.add_grade("efg12h", 5)
        gb.add_grade("abcd12", 5)
        self.assertEqual({("Arpita Ekka", "efg12h"): 3, ("Tamara Takács", "abcd12"): 1}, gb.get_worst_recent_grade())
        gb.drop_student("abcd12")
        self.assertEqual({("Arpita Ekka", "efg12h"): 3}, gb.get_worst_recent_grade())

    def test_get_student_codes_with_many_5s(self):
        gb = Gradebook()
        self.assertEqual([], gb.get_student_codes_with_many_5s())
        students1 = [("Ariona Pirku", "efg12h"), ("Emmett Taylor Elliott", "abcd12")]
        gb.add_students(students1)
        self.assertEqual([], gb.get_student_codes_with_many_5s())
        gb.add_grade("efg12h", 2)
        gb.add_grade("efg12h", 4)
        self.assertEqual([], gb.get_student_codes_with_many_5s())
        gb.add_grade("abcd12", 5)
        self.assertEqual(["abcd12"], gb.get_student_codes_with_many_5s())
        gb.add_grade("efg12h", 5)
        gb.add_grade("efg12h", 5)
        self.assertEqual(["abcd12"], gb.get_student_codes_with_many_5s())
        gb.add_grade("efg12h", 5)
        self.assertEqual(["abcd12", "efg12h"], sorted(gb.get_student_codes_with_many_5s()))
        gb.drop_student("abcd12")
        self.assertEqual(["efg12h"], gb.get_student_codes_with_many_5s())
        gb.add_grade("efg12h", 2)
        self.assertEqual([], gb.get_student_codes_with_many_5s())

def suite():
    suite = unittest.TestSuite()
    testfuns = ["test_add_student", "test_add_students", "test_drop_student",
                "test_get_student_name", "test_get_all_student_neptun_codes", "test_get_all_student_names",
                "test_search_students_by_name", "test_add_grade",
                "test_get_student_grades", "test_get_number_of_students_with_low_grades",
                "test_get_average_grade", "test_get_avarage_grade_per_student",
                "test_get_worst_recent_grade","test_get_student_codes_with_many_5s"]
    [suite.addTest(TestGradebook(fun)) for fun in testfuns]
    return suite

runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite())
```
