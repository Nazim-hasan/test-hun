## Using Google Colaboratory


**Version used in this notebook: Python 3.6**

**Set Google Colab to the correct version:** Edit --> Notebook Settings --> Runtime type --> Python 3

Google Colab edits Jupyter notebooks with the `.ipynb` extension. A Jupyter notebook consists of code cells and text cells. Code cells can be executed individually, but the whole notebook can also be run at once.

**Run a single cell:** using the button that appears in the upper-left corner of the cell when you click it (or by selecting the cell and pressing Ctrl+Enter).

**Run the whole notebook:** from the Runtime menu using **Run all** (or Ctrl+F9).

If a cell produces any displayable output (explicit use of `print`, plotting, or the value of the last standalone expression), it will appear below the cell. If an error occurs during the execution of a cell, the error is printed and notebook execution stops at that point.

Execution is provided by the runtime automatically attached to the notebook; its status is visible in the upper-right corner, e.g. "Connected".

As in Google Docs, a notebook can be shared with **view only** and **can edit** permissions. If a notebook is shared with someone with **view only** permission, they cannot edit or run it; they only see the text and code cells and the output from the last run under the code cells. Users with **view only** permission can, of course, copy the notebook to their own Drive and then edit and run it there (File --> Save a copy in Drive). In addition, a **view only** notebook can also be opened in **playground mode**: in this case a temporary copy is created that can be edited, but if you do not save it to your Drive before leaving, your changes will be lost.

Notebooks can be downloaded and used with a local Jupyter runtime as well. The `.ipynb` files can be converted to `.html` files using Jupyter, so they can be read offline without a runtime.

**Running a Jupyter notebook**

Jupyter notebooks work similarly to the Python interpreter. After running the notebook, variables declared in the notebook's code cells persist as long as the runtime is connected. If you later re-run a cell (that has already been run), its result may differ from what you got when you first ran the whole notebook. For example: you define a variable with an initial value in one cell, then increase it in the next cell. If the latter runs twice, the variable's value will be different than if it ran once. The **Restart runtime** option in the Runtime menu **restarts the runtime**, helping you avoid the issue described above.

**Google Colab runtime settings:**

Google also provides a GPU for free to speed up notebooks. However, GPU acceleration only affects libraries that can use it, such as TensorFlow or PyTorch. For simple Python/Numpy code, this setting does not change runtime in Colab. GPU acceleration can be enabled via: Edit --> Notebook Settings --> Hardware accelerator --> GPU.

## Running Python scripts

Although during the course we will only use the interpreter-like Google Colaboratory, the most basic way to use Python is to run Python script files. Script files conventionally use the `.py` file extension. Running them requires a Python environment installed on your own (or a remote) machine. On Windows, WinPython and Anaconda distributions are the most common, and the latter is recommended. On Linux, Python 2.7 is typically part of the system, but Python 3 can be installed via the distribution's package manager (on Ubuntu, for example, `apt-get`). On Windows you can run a script from `cmd` or from the Anaconda prompt; on Linux from `bash`, e.g.:

```
python3 script.py
```

The script runs and then the Python process terminates: unlike an interpreter environment (e.g. Jupyter/Google Colab), the defined variables do not persist.

**Variables in Python:**
In Python, variables are dynamically typed and no explicit type declaration is required. A variable always stores the memory address of an object, so the variable's type is the type of the object it currently references. The required memory for a new object is allocated automatically, so there is no need for explicit declarations or manual memory allocation.

## Lists

A list can contain arbitrary objects of mixed types, even other lists.

Documentation: https://docs.python.org/3/tutorial/datastructures.html

**Characteristics of Python lists:**

In the default Python implementation (CPython), a list is implemented as an array of pointers written in C. Its size grows dynamically: when the list size reaches the currently allocated capacity, a new, larger memory area is allocated and the old pointers are copied into the new area.

Since the list is implemented as an array, accessing and overwriting an element is **O(1)**, i.e. constant time. Appending to the end (`append(), extend()`) and removing from the end (`pop(-1)`) is also constant time, except when the allocated capacity is exhausted and the list has to be reallocated. Deleting or inserting an element in the middle (or at the beginning) is **O(n)**, since this is not a linked list: elements after the insertion/deletion point must be shifted left or right.


````python
my_list = [1, "abc", None, False]
print("A simple list. Elements may have different types:", my_list)

empty_list = []
print("An empty list:", empty_list, '\n')

copied_list = list(my_list)
print("A copy of my_list:", copied_list)
print("Is copied_list the same object as my_list? ", copied_list is my_list)

jagged_list = [[2,3], ['r',[]], 'a', [None]]
print("Lists can contain any type of objects, other lists as well: ", jagged_list)
````

````python
print("\nIndexing a list: ")
basic_list = [1,2,3,4,5]
print("A basic list:", basic_list)
print("Element at index 2 in the basic list (third element):", basic_list[2])
                                  # Indexing starts with ZERO and goes to length-1

print("2nd element counting from the back of the basic list:", basic_list[-2])
                                  # Negative index ''-k' corresponds to length-k

length = len(jagged_list) # this is a function returning an int, the length of the given list
print(length, 'is the length of the jagged list: ', jagged_list)
````

````python
print("\nMethods of the list class:")

basic_list.append(3)
print("Appended one element to the end: ", basic_list)

basic_list.append([3, 4])
print("Appended one element (a list with a length of two) to the end: ", basic_list)

basic_list.extend([3, 4])
print("Extended the list with two elements: ", basic_list)

basic_list.insert(0, None)
print("Inserted an element to the from of the list (index 0): ", basic_list)

basic_list.insert(-1, None)
print("Inserted an element to the back of the list (index -1): ", basic_list)
````

Other list methods:

`pop(), remove(), clear(), sort(), ...`

````python

print("\nSlice and stride: ")
# Format of list slicing: my_list[from:to:step]
#   iteration goes from 'from' (inclusive) to 'to' (exclusive) with a step size of 'step'
#   second colon (':') and any of 'from', 'to' or 'step' can be omitted

my_list = [0,1,2,3,4,5]
print("The original list: ", my_list)
print("From idx 2 to idx 4 (exclusive): ", my_list[2:4])
print("From idx 2 to idx 4 (exclusive) with step size 2: ", my_list[2:4:2])
print("From idx 2 to the end: ", my_list[2:])
print("From the beginning till idx 2 (exclusive): ", my_list[:2], '\n')

print("Whole list with step size 2: ", my_list[::2])
print("Reversed list: ", my_list[::-1])
print("Reversed list with step size 2: ", my_list[::-2])
print("Reversed list with step size 2, starting"\
            " from second but last element towards the beginning: ", my_list[-2::-2])

# when slicing the reversed list, the 'from' marks the latter index and
# 'to' is the smaller one, closer to the front as we iterate in the opposite direction
````

````python
print("When slicing, out of range indices will not raise an error: ", my_list[:8])
print("When slicing, out of range indices will not raise an error: ", my_list[7:8])

# indexing with a single out of range index would of course cause an error

print("\nString is also a sequence type:")
s = 'abcdef'
print(s[2])
print(s[-1])
print(s[:4])
print(s[-2:1:-1])
````

## Conditionals, loops, the `pass` keyword

In Python, blocks are indicated using whitespace: after `for`, `while`, `if`, `else`, `elif`, an indented block must follow. Indentation means increasing the number of whitespace characters at the beginning of the line. Every statement within a block must start with the same number of whitespace characters. Whitespace can be a tab or spaces, but mixing the two is not allowed (Google Colab replaces tabs with spaces).


````python

print("\nIf statements: ")
ch = 'a'
if ch == 'a':
  print(1)

if ch == 'b':
  print(2)
elif ch == 'c':
  print(3)

if ch == 'b':
  print(4)
else:
  print(5)

if ch == 'd':
  print(6)
elif ch == 'a':
  print(7)
else:
  print(8)
````

````python
print("\nFor loop with the range() built-in function: ")
for i in range(2,7,2): # simple ranged for iterating through the range object
    print('  ', i)

sum = 0
for k in range(20): # summation with for
    sum += k
print("\nSum is ", sum)

# range() is a built-in function which returns an iterator over integers
#   range(to) will iterate from 0 (inclusive) until 'to' (exclusive)
#   range(from, to) will iterate from 'from' (inclusive) until 'to' (exclusive)
#   range(from, to, step), similar as in list slicing

# Warning! In Python 2.x range() returned a list, not an iterator.

print("\nIterating over elements of a list: ")
my_list = ['hello', None, '22', 2]
for item in my_list:
    print('  ', item)

````

````python
print("\nWhile loop: ")
i = 0
while(i < 5):
    print('  ', i)
    i +=1

print('\nPass is the empty statement ')
for i in range(10):
    pass    # do nothing: empty indented block results in a syntax error
            #    this is why 'pass' exists
````
