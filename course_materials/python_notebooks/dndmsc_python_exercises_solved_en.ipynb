ReLU(z) = max(0, z)

For example:

Define the `mult_table` function, which expects a positive integer `n` as a parameter. The function should create and return the multiplication table from 0 to n as a nested list. If you index the returned nested list with i and j, you should get the value i*j.

For example:
```
mult_table(0) -> [[0]]
mult_table(1) -> [[0, 0], [0, 1]]
mult_table(2) -> [[0, 0, 0], [0, 1, 2], [0, 2, 4]]

mult_table(6)[4][5] -> 20
```


````python
# place of solution

def mult_table(n):
  mtable = []
  for i in range(n+1):
    row = []
    mtable.append(row)
    for j in range(n+1):
      row.append(i*j)
  return mtable

# Tests
assert mult_table(0) == [[0]]
assert mult_table(1) == [[0, 0], [0, 1]]
assert mult_table(2) == [[0, 0, 0], [0, 1, 2], [0, 2, 4]]
assert mult_table(6)[4][5] == mult_table(6)[5][4] == 20
assert mult_table(10)[0][10] == 0
assert mult_table(10)[10][10] == 100
print("Tests were successful.")
````

## **Exercise 5**

Define the `fibonacci` function, which expects a non-negative integer `n` as a parameter. The function should compute and return the n-th element of the Fibonacci sequence.

For example:
```
fibonacci(0) -> 0
fibonacci(1) -> 1
fibonacci(n) -> fibonacci(n-1) + fibonacci(n-2)
```


````python
# place of solution

# EFFICIENT SOLUTION: O(n) (linear) complexity
def fibonacci(n):
  series = [0,1]
  for i in range(2,n+1):
    series.append(series[-1] + series[-2])
  return series[n]

# Tests
assert fibonacci(0) == 0
assert fibonacci(1) == 1
for test_i in range(2,10):
  assert fibonacci(test_i) == fibonacci(test_i-1) + fibonacci(test_i-2)
print("Tests were successful.")
````

````python
# place of solution

# INEFFICIENT SOLUTION: O(2^n) (exponential) instead of O(n) (linear) complexity
def fibonacci(n):
  return fibonacci(n-1) + fibonacci(n-2) if n >= 2 else n

# Tests
assert fibonacci(0) == 0
assert fibonacci(1) == 1
for test_i in range(2,10):
  assert fibonacci(test_i) == fibonacci(test_i-1) + fibonacci(test_i-2)
print("Tests were successful.")
````

````python
# place of solution

# INEFFICIENT SOLUTION: O(n^2) instead of O(n) complexity
fibonacci = lambda n: fibonacci(n-1) + fibonacci(n-2) if n >= 2 else n

# Tests
assert fibonacci(0) == 0
assert fibonacci(1) == 1
for test_i in range(2,10):
  assert fibonacci(test_i) == fibonacci(test_i-1) + fibonacci(test_i-2)
print("Tests were successful.")
````

## **Exercise 6**

Define the `is_paren_correct` function, which, for two types of opening and closing brackets (`()` and `[]`), determines whether the bracket order is correct in the input string expression.

For example:
```
is_paren_correct("") -> True
is_paren_correct("([]())[[([]())]]") -> True
is_paren_correct("([)]") -> False
is_paren_correct("(") -> False
```


````python
# place of solution

def is_paren_correct(expr):
  pstack = []
  for ch in expr:
    if ch == ')' and (len(pstack) == 0 or pstack[-1] != '('):
      return False
    elif ch == ']' and (len(pstack) == 0 or pstack[-1] != '['):
      return False
    elif ch == ']' or ch == ')':
      pstack.pop()
    elif ch == '[' or ch == '(':
      pstack.append(ch)
  return len(pstack) == 0   # return True if empty

# Tests
assert is_paren_correct("")
assert is_paren_correct("()")
assert is_paren_correct("[]")
assert is_paren_correct("([]())[[([]())]]")
assert not is_paren_correct("([)]")
assert not is_paren_correct("(")
assert not is_paren_correct(")")
assert not is_paren_correct("[)")
assert not is_paren_correct("(]")
print("Tests were successful.")
````

## **Exercise 7**

Define the `triangle_area` function, which takes a list of triples of numbers and returns, for each triple, whether a triangle can be formed from the given side lengths and, if so, the area of that triangle. The return value is a list whose i-th element is the area of the triangle formed from the i-th triple in the input list; if a triangle cannot be formed, the element should be `None`.

To compute the area of a triangle, you can use Heron's formula. A triangle can be formed from a triple only if every side length is positive and the triangle inequality holds.

For example:
```
triangle_area([]) -> []
triangle_area([(5,3,4), (4,1,2)]) -> [6, None]
```


````python
import math

# place of solution

def triangle_area(triples):
  areas = []
  for a, b, c in triples:
    if (a >= b+c) or (b >= a+c) or (c >= a+b):
      areas.append(None)
    else:
      s = (a+b+c)/2.
      areas.append(math.sqrt(s*(s-a)*(s-b)*(s-c)))
  return areas

# Tests
assert triangle_area([]) == []
test_item1, test_item2 = triangle_area([(5,4,3), (4,1,2)])
assert (test_item1 - 6.) < 0.0001
assert test_item2 is None
print("Tests were successful.")
````

Define the `sort_by_age` function, which takes as input a dictionary of name: age pairs and returns the list of names sorted by age in decreasing order.

For example:
```
sort_by_age({}) -> []
sort_by_age({'Anett': 32, 'Barna': 25, 'Cecil':78, 'Dorina':45}) -> ['Cecil', 'Dorina', 'Anett', 'Barna']
```


````python
# place of solution

def sort_by_age(age_dict):
  dict_items = age_dict.items()
  items_sorted_by_age = sorted(dict_items, key=lambda name_age_pair: name_age_pair[1])[::-1]
                                                  # [::-1] reverses the list
  age_list = []
  for name, age in items_sorted_by_age:
    age_list.append(name)
  return age_list

# Tests
assert sort_by_age({}) == []
assert sort_by_age({'Anett': 32, 'Barna': 25, 'Cecil':78, 'Dorina':45}) ==\
                                           ['Cecil', 'Dorina', 'Anett', 'Barna']
print("Tests were successful.")
````

## **Exercise 9**

Define the `remaining_str` function, which takes a string and a list of number pairs as input. The first element of each pair is the index of the first character to remove, and the second element is the index of the first character after the removed part. Assume that all pairs refer to valid indices in the string.

For example:
```
remaining_str("abcdef", []) -> "abcdef"
remaining_str("abcdef", [(0,3)]) -> "def"
remaining_str("abcdef", [(0,3), (4,5)]) -> "df"
remaining_str("abcdef", [(2,5), (0,1)]) -> "bf"
```


````python
# place of solution

def remaining_str(my_str, intervals_to_cut):
  remaining_idxs = [True]*len(my_str)
  for interval_start, interval_end in intervals_to_cut:
    for idx in range(interval_start, interval_end):
      remaining_idxs[idx] = False

  remaining_chars = []
  for idx in range(len(my_str)):
    if remaining_idxs[idx]:
      remaining_chars.append(my_str[idx])
  return ''.join(remaining_chars)


# Tests
assert remaining_str("abcdef", []) == "abcdef"
assert remaining_str("abcdef", [(0,3)]) == "def"
assert remaining_str("abcdef", [(0,3), (4,5)]) == "df"
assert remaining_str("abcdef", [(2,5), (0,1)]) == "bf"
print("Tests were successful.")
````

They store each person's available times in a dictionary in the form name: list of numbers. Define the `acceptable_times` function, which returns the list of times that work for everyone based on the given dictionaries. If there is no such time, it should return an empty list. For simplicity, each time slot is represented by a single number.

For example:
```
acceptable_times({}) -> []
acceptable_times({'Anna': [5,7,3,2]}) -> [2,3,5,7]
acceptable_times({'Anett':[3,5,6], "Barna":[1,3,2,6], "Cecil":[7,5,3,6], "Dorina":[2,1,3,5,6,9]}) -> [3,6]
acceptable_times({'Anett':[3,5,6], "Barna":[1,3,2,9], "Cecil":[7,5,3,6], "Dorina":[2,1,5,6,9]}) -> []
```


````python
# place of solution

def acceptable_times(time_dict):
  if len(time_dict) == 0:
    return []
  else:
    dict_timelists = list(time_dict.values())
    acceptable_set = set(dict_timelists[0])
    for time_list in dict_timelists[1:]:
      acceptable_set = acceptable_set.intersection(set(time_list))
    return list(acceptable_set)

# Tests
assert acceptable_times({}) == []
assert acceptable_times({'Anett':[3,5,6], "Barna":[1,3,2,6], "Cecil":[7,5,3,6], "Dorina":[2,1,3,5,6,9]}) == [3,6]
assert acceptable_times({'Anett':[3,5,6], "Barna":[1,3,2,9], "Cecil":[7,5,3,6], "Dorina":[2,1,5,6,9]}) == []
print("Tests were successful.")
````

## **Exercise 11**

Define the `same_magnitude` function, which expects a list of positive numbers as a parameter. It should return `True` if every number in the list is greater than half of the maximum element. If any number is not positive, the function should return `False`.

For example:
```
same_magnitude([2,5,3,2]) -> False
same_magnitude([4,5,3,4]) -> True
same_magnitude([4,5,3,-2]) -> False
same_magnitude([0,0,0]) -> False
same_magnitude([-2,-2,-2]) -> False
```


````python
# place of solution

def same_magnitude(my_list):
  limit = max(my_list)*.5
  for item in my_list:
    if (item <= 0) or (item <= limit):
      return False
  return True

# Tests
assert not same_magnitude([2,5,3,2])
assert same_magnitude([4,5,3,4])
assert not same_magnitude([4,5,3,-2])
assert not same_magnitude([0,0,0])
assert not same_magnitude([-2,-2,-2])
print("Tests were successful.")
````
